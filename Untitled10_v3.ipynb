### 2.1 路径 / IO / 标签解析

标签解析同时吃 AABB（4 坐标）和 OBB（8 坐标），统一输出归一化 AABB —— 这是 v1 最大的隐性 bug 来源。

````python
IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tif", ".tiff"}
_SPLIT_ALIASES = {"train": "train", "training": "train",
                  "valid": "valid", "val": "valid", "validation": "valid",
                  "test": "test"}
_SKIP_DIR_NAMES = {"images", "image", "labels", "label", "annotations",
                   "annotations_txt", "jsons", "xmls", "train", "valid", "val", "test"}


def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


def ensure_dir(path):
    path = Path(path)
    path.mkdir(parents=True, exist_ok=True)
    return path


def load_yaml(path):
    with open(path, "r", encoding="utf-8") as f:
        return yaml.safe_load(f) or {}


def yolo_class_names(data_yaml):
    names = data_yaml.get("names", [])
    if isinstance(names, dict):
        return [names[k] for k in sorted(names, key=lambda x: int(x))]
    return list(names)


def iter_images(directory):
    directory = Path(directory)
    if not directory.exists():
        return
    for p in sorted(directory.iterdir()):
        if p.is_file() and p.suffix.lower() in IMAGE_EXTS:
            yield p


def parse_yolo_line(line):
    # -> (cls_id, x1, y1, x2, y2) 归一化 AABB；非法行返回 None
    parts = line.split()
    if len(parts) < 5:
        return None
    try:
        cls_id = int(float(parts[0]))
        c = [float(v) for v in parts[1:]]
    except (ValueError, TypeError):
        return None
    if len(c) == 4:
        cx, cy, w, h = c
    elif len(c) >= 8:                       # OBB 四点 -> 外接矩形
        xs, ys = c[0::2], c[1::2]
        x1, y1, x2, y2 = min(xs), min(ys), max(xs), max(ys)
        cx, cy, w, h = (x1 + x2) / 2, (y1 + y2) / 2, x2 - x1, y2 - y1
    else:
        return None
    return (cls_id, cx - w / 2, cy - h / 2, cx + w / 2, cy + h / 2)


def read_label_boxes(label_path):
    label_path = Path(label_path)
    if not label_path.exists():
        return []
    out = []
    for line in label_path.read_text(encoding="utf-8", errors="ignore").splitlines():
        r = parse_yolo_line(line)
        if r is not None:
            out.append(r)
    return out


def find_split_dirs(root):
    # -> {"train": Path, "valid": Path, "test": Path}，只保留真的有 images/ 的
    root = Path(root)
    out = {}
    if not root.exists():
        return out
    for d in [root] + [p for p in root.iterdir() if p.is_dir()]:
        key = _SPLIT_ALIASES.get(d.name.lower())
        if key and (d / "images").is_dir():
            out.setdefault(key, d)
    return out


def is_detection_dataset(root):
    return bool(find_split_dirs(root))


def find_leaf_class_dirs(root):
    # 找"直接包含图片"的目录作为类别目录，跳过 images/labels/train 这类结构目录
    root = Path(root)
    if not root.exists():
        return []
    found = []
    for d in [root] + [p for p in root.rglob("*") if p.is_dir()]:
        if d.name.lower() in _SKIP_DIR_NAMES:
            continue
        if (d / "images").is_dir() or (d / "labels").is_dir():
            continue
        try:
            if any(p.is_file() and p.suffix.lower() in IMAGE_EXTS for p in d.iterdir()):
                found.append(d)
        except OSError:
            continue
    return sorted(set(found))


def register_classification_root(name, root):
    root = Path(root)
    if not root.exists():
        print("[skip] %s: 路径不存在 -> %s" % (name, root))
        return
    rp = str(root.resolve())
    for _, existing in CLASSIFICATION_ROOTS:
        if str(Path(existing).resolve()) == rp:
            return
    CLASSIFICATION_ROOTS.append((name, root))
    print("[registered] %s: %s" % (name, root))


def find_class_names_file(root):
    # 在 root 下最多 3 层内找 data.yaml / classes.txt / *.json 之类的类名文件
    cands = {"data.yaml", "data.yml", "classes.txt", "class_names.txt",
             "names.txt", "classes.json", "class_names.json", "classes.csv"}
    level = [Path(root)]
    for _ in range(3):
        nxt = []
        for d in level:
            try:
                for p in d.iterdir():
                    if p.is_dir():
                        nxt.append(p)
                        continue
                    if p.name.lower() not in cands:
                        continue
                    try:
                        if p.suffix.lower() in {".yaml", ".yml"}:
                            n = yolo_class_names(load_yaml(p))
                            if n:
                                return n
                        elif p.suffix.lower() == ".json":
                            obj = json.loads(p.read_text(encoding="utf-8", errors="ignore"))
                            if isinstance(obj, dict) and "names" in obj:
                                n = yolo_class_names(obj)
                                if n:
                                    return n
                            if isinstance(obj, dict) and obj and all(str(k).isdigit() for k in obj):
                                return [obj[k] for k in sorted(obj, key=lambda x: int(x))]
                            if isinstance(obj, list) and obj and isinstance(obj[0], str):
                                return obj
                        else:
                            n = [l.strip() for l in p.read_text(
                                encoding="utf-8", errors="ignore").splitlines() if l.strip()]
                            if n:
                                return n
                    except Exception:
                        continue
            except OSError:
                continue
        level = nxt
    return None


def normalize_brand_name(name):
    return re.sub(r"[^a-z0-9]+", "", str(name).lower())


def sanitize_name(name):
    name = re.sub(r"[^A-Za-z0-9._ -]+", "_", str(name)).strip()
    return name or "unknown"


def autocast_ctx(enabled):
    if not enabled:
        return nullcontext()
    try:
        return torch.amp.autocast("cuda", dtype=torch.float16)
    except Exception:
        return torch.cuda.amp.autocast()


def make_grad_scaler(enabled):
    try:
        return torch.amp.GradScaler("cuda", enabled=enabled)
    except (AttributeError, TypeError):
        return torch.cuda.amp.GradScaler(enabled=enabled)


set_seed(SEED)
print("IO / label utils ready.")
````

### 2.2 几何：containment / 软匹配

v1 用 `containment >= 1.0`（logo 必须 100% 落在 player 框内）做硬过滤。
但 player 检测器框的是"可见躯干"，被裁切/遮挡的球员框很小，logo 又常在短裤、袖口、人身轮廓边缘——
一个像素出框就整条丢弃，实测丢掉了 17% 已经检测正确的 logo。

v2 改成软匹配：先外扩 player 框 10%，再要求 `面积包含率 >= 0.55`，或者退化条件是"logo 中心点落在人框内"。
这个判定同时输出成一个**分数**（`containment`），后续可以当特征用，而不是一刀切。

````python
def containment_ratio(inner_poly, outer_poly):
    # inner 凸多边形落在 outer 内的面积比例，裁剪到 [0, 1]（修掉 v1 的 1.0000000172 抖动）
    inner = np.asarray(inner_poly, dtype=np.float32).reshape(-1, 2)
    outer = np.asarray(outer_poly, dtype=np.float32).reshape(-1, 2)
    area = abs(cv2.contourArea(inner))
    if area <= 1e-6:
        return 0.0
    inter, _ = cv2.intersectConvexConvex(inner, outer)
    return float(min(1.0, max(0.0, inter / area)))


def expand_poly(poly, pad=0.10):
    x1, y1 = poly.min(axis=0)
    x2, y2 = poly.max(axis=0)
    w, h = max(x2 - x1, 1e-6), max(y2 - y1, 1e-6)
    return np.array([[x1 - pad * w, y1 - pad * h],
                     [x2 + pad * w, y1 - pad * h],
                     [x2 + pad * w, y2 + pad * h],
                     [x1 - pad * w, y2 + pad * h]], dtype=np.float32)


def logo_player_match(logo_poly, player_polys,
                      pad=CONTAINMENT_PAD, area_thr=CONTAINMENT_AREA_THR):
    # -> (是否保留, 最佳包含率, 中心点是否落在某个 player 框内)
    best, center_in = 0.0, False
    c = logo_poly.mean(axis=0)
    for p in player_polys:
        pe = expand_poly(p, pad)
        best = max(best, containment_ratio(logo_poly, pe))
        x1, y1 = pe.min(axis=0)
        x2, y2 = pe.max(axis=0)
        if x1 <= c[0] <= x2 and y1 <= c[1] <= y2:
            center_in = True
    ok = (best >= area_thr) or (center_in and best >= 0.30)
    return ok, best, center_in


def result_polygons(result):
    # 兼容 Ultralytics 的 OBB 与普通检测框，统一返回 polygon + confidence
    obb = getattr(result, "obb", None)
    if obb is not None and len(obb) > 0:
        polys = obb.xyxyxyxy.cpu().numpy().reshape(-1, 4, 2)
        return [p.astype(np.float32) for p in polys], obb.conf.cpu().numpy()
    boxes = getattr(result, "boxes", None)
    if boxes is not None and len(boxes) > 0:
        polys, confs = [], []
        for b, c in zip(boxes.xyxy.cpu().numpy(), boxes.conf.cpu().numpy()):
            x1, y1, x2, y2 = b
            polys.append(np.array([[x1, y1], [x2, y1], [x2, y2], [x1, y2]], dtype=np.float32))
            confs.append(float(c))
        return polys, np.asarray(confs, dtype=np.float32)
    return [], np.asarray([], dtype=np.float32)


def poly_to_xyxy(poly, w, h):
    x1, y1 = poly.min(axis=0)
    x2, y2 = poly.max(axis=0)
    return (max(0, int(np.floor(x1))), max(0, int(np.floor(y1))),
            min(w, int(np.ceil(x2))), min(h, int(np.ceil(y2))))


def poly_with_context(poly, w, h, ctx=CROP_CONTEXT):
    # 送进 Doctor 之前把框外扩一点，保留球衣纹理（品牌判别很强的线索）
    x1, y1 = poly.min(axis=0)
    x2, y2 = poly.max(axis=0)
    dx, dy = (x2 - x1) * ctx, (y2 - y1) * ctx
    return (max(0, int(np.floor(x1 - dx))), max(0, int(np.floor(y1 - dy))),
            min(w, int(np.ceil(x2 + dx))), min(h, int(np.ceil(y2 + dy))))


def iou_xyxy(a, b):
    ix1, iy1 = max(a[0], b[0]), max(a[1], b[1])
    ix2, iy2 = min(a[2], b[2]), min(a[3], b[3])
    iw, ih = max(0.0, ix2 - ix1), max(0.0, iy2 - iy1)
    inter = iw * ih
    aa = max(0.0, a[2] - a[0]) * max(0.0, a[3] - a[1])
    bb = max(0.0, b[2] - b[0]) * max(0.0, b[3] - b[1])
    return inter / (aa + bb - inter + 1e-9)


print("Geometry utils ready.")
````

### 3.5 定义 unseen-brand holdout（必须最先做，否则泛化改了什么都不知道）

现在在 Sponsor-detector-2 上评测，那里的品牌检测器全都见过，所以**你永远看不到"没见过"的失败**。

这一格从 LogoDet-3K 里整类挑出 150 个品牌作为 holdout：

- 它们的 crop **不喂给 Doctor**（3.4/4.2 抽 crop 时跳过）；
- 它们的图片 **不并入 logo detector 训练集**（4.4 合并时跳过）；
- 只在 10.x 的 unseen-brand recall 评测里出现。

这样你才有两栏可比的数字：`recall@seen` vs `recall@unseen`。

````python
GENERIC_CLASS_NAMES = {"logo", "sponsor", "brand", "logoimage", "0", "class0", ""}

EXCLUDE_BRAND_KEYS = set()        # 归一化后的品牌 key，抽 crop / 建词表时要跳过
UNSEEN_BRAND_NAMES = []
UNSEEN_CLS_IDS = {}               # {dataset_name: set(cls_id)}

if EXTERNAL_DETECTION_DATASETS:
    _ds_name, _ds_root = EXTERNAL_DETECTION_DATASETS[0]
    _names = None
    if _ds_name in EXTERNAL_CLASS_NAMES_FILES:
        _names = [l.strip() for l in Path(EXTERNAL_CLASS_NAMES_FILES[_ds_name]).read_text(
            encoding="utf-8", errors="ignore").splitlines() if l.strip()]
    if _names is None:
        _names = find_class_names_file(_ds_root)

    if not _names:
        print("!! %s 找不到类别名文件，无法定义 unseen-brand holdout" % _ds_name)
    else:
        _cand = [(i, n) for i, n in enumerate(_names)
                 if normalize_brand_name(n) not in GENERIC_CLASS_NAMES]
        _rng = random.Random(UNSEEN_HOLDOUT_SEED)
        # 安全上限：holdout 不超过总类数的 5%，否则会连带把训练集掏空
        _n_pick = min(UNSEEN_HOLDOUT_N, max(3, int(0.05 * len(_cand))), len(_cand))
        _pick = _rng.sample(_cand, _n_pick)
        UNSEEN_BRAND_NAMES = [n for _, n in _pick]
        UNSEEN_CLS_IDS[_ds_name] = {i for i, _ in _pick}
        EXCLUDE_BRAND_KEYS = {normalize_brand_name(n) for n in UNSEEN_BRAND_NAMES}
        print("unseen holdout: %d 个品牌（来自 %s），例如: %s"
              % (len(UNSEEN_BRAND_NAMES), _ds_name, UNSEEN_BRAND_NAMES[:8]))
else:
    print("没有外部检测集，跳过 unseen holdout 定义（10.x 的泛化评测会跳过）")
````

## 4. 检测器数据整理

### 4.1 从多类检测标签里抽品牌 crop（给 Doctor 用）

必须在 class-agnostic 转换**之前**做，否则品牌名就丢了。
泛化类别名（`logo` / `sponsor` / `brand` / `0` …）会被跳过，因为对分类没有信息量。

````python

def extract_brand_crops(src_root, out_root, prefix, names=None,
                        max_crops=None, min_px=8, exclude_keys=None):
    """从 YOLO 多类检测集抠出 logo crop，按品牌名归档。返回统计 dict。"""
    src_root = Path(src_root)
    out_root = ensure_dir(out_root)
    if names is None:
        names = yolo_class_names(load_yaml(src_root / "data.yaml")) if (src_root / "data.yaml").exists() else []

    stats = Counter()
    for split_key, split_dir in find_split_dirs(src_root).items():
        img_dir, lbl_dir = split_dir / "images", split_dir / "labels"
        for img_path in iter_images(img_dir):
            stats["images"] += 1
            label_path = lbl_dir / (img_path.stem + ".txt")
            if not label_path.exists():
                continue
            boxes = read_label_boxes(label_path)
            if not boxes:
                continue
            image = cv2.imread(str(img_path))
            if image is None:
                stats["bad_image"] += 1
                continue
            h, w = image.shape[:2]

            for box_idx, (cls_id, x1n, y1n, x2n, y2n) in enumerate(boxes):
                stats["boxes"] += 1
                brand = names[cls_id] if 0 <= cls_id < len(names) else "class_%d" % cls_id
                _key = normalize_brand_name(brand)
                if _key in GENERIC_CLASS_NAMES:
                    stats["skipped_generic"] += 1
                    continue
                if exclude_keys and _key in exclude_keys:
                    stats["skipped_holdout"] += 1
                    continue

                x1 = max(0, int(x1n * w)); y1 = max(0, int(y1n * h))
                x2 = min(w, int(x2n * w)); y2 = min(h, int(y2n * h))
                if x2 - x1 < min_px or y2 - y1 < min_px:
                    stats["skipped_tiny"] += 1
                    continue

                brand_dir = ensure_dir(out_root / sanitize_name(brand))
                crop_name = "%s_%s_%s_%d.jpg" % (prefix, split_key, img_path.stem, box_idx)
                cv2.imwrite(str(brand_dir / crop_name), image[y1:y2, x1:x2])
                stats["crops"] += 1
                if max_crops is not None and stats["crops"] >= max_crops:
                    return stats
    return stats


print("extract_brand_crops ready.")
````

### 4.3 自动生成 `other` 负类

Doctor 是闭集分类器，v1 的后果是：任何没见过的品牌都会被硬塞进 141 个类里，还给出 0.999 的置信度。

这里用球员数据集里的转播画面，**在人体框之外**随机裁背景块（广告牌、看台、草皮、角旗）——
它们正是最容易被误判成"球衣赞助"的东西。把这些块标成 `other` 类，Doctor 就学会了"这不是品牌"。

````python
def build_other_class_crops(player_root, out_dir, per_image=2,
                            max_images=1500, min_px=32, seed=SEED):
    """在人体框之外随机裁背景块，作为 other 负类。"""
    rng = random.Random(seed)
    out_root = ensure_dir(Path(out_dir) / "other")
    splits = find_split_dirs(player_root)
    split_dir = splits.get("train") or next(iter(splits.values()), None)
    if split_dir is None:
        print("没有找到 player 训练集，跳过 other 负类生成")
        return 0

    img_dir, lbl_dir = split_dir / "images", split_dir / "labels"
    images = list(iter_images(img_dir))[:max_images]
    n = 0
    for img_path in images:
        image = cv2.imread(str(img_path))
        if image is None:
            continue
        h, w = image.shape[:2]
        person_boxes = [(x1 * w, y1 * h, x2 * w, y2 * h)
                        for _, x1, y1, x2, y2 in read_label_boxes(lbl_dir / (img_path.stem + ".txt"))]

        made = 0
        for _ in range(per_image * 8):
            if made >= per_image:
                break
            bw = rng.uniform(0.10, 0.30) * w
            bh = bw * rng.uniform(0.6, 1.6)
            if bh > 0.5 * h:
                continue
            x1 = rng.uniform(0, w - bw)
            y1 = rng.uniform(0, h - bh)
            box = (x1, y1, x1 + bw, y1 + bh)
            if any(iou_xyxy(box, pb) > 0.0 for pb in person_boxes):
                continue                      # 和人体框有重叠 -> 可能是球衣，丢掉
            crop = image[int(y1):int(y1 + bh), int(x1):int(x1 + bw)]
            if crop.shape[0] < min_px or crop.shape[1] < min_px:
                continue
            cv2.imwrite(str(out_root / ("other_%s_%d.jpg" % (img_path.stem, made))), crop)
            made += 1
            n += 1
    return n


if ADD_AUTO_OTHER_CLASS:
    if OTHER_CROPS.exists():
        shutil.rmtree(OTHER_CROPS)
    n_other = build_other_class_crops(PLAYER_DATASET_DIR, OTHER_CROPS,
                                      per_image=AUTO_OTHER_PER_IMAGE,
                                      max_images=AUTO_OTHER_MAX_IMAGES)
    print("生成 other 负类 crop:", n_other)
else:
    print("跳过 other 负类生成")
````

### 4.4 融合 Logo detection 数据集 → class-agnostic 副本

v1 的关键 bug：注释写着"OBB 只改 class，坐标原样保留"，于是同一个数据集里混着 4 坐标和 8 坐标两种标签。
Ultralytics 的 detect 模型每行只认 5 个数，8 坐标的行会被当作损坏样本**整行丢掉**——
这就是 `logo_detection_e92p3` 那 1050 个框几乎没学到东西的原因。

v2 统一转成 `cls cx cy w h`，并把丢弃行数打出来。
同时把 4.3 生成的负样本图（空标签）插进 train split，让检测器学会"广告牌不是球衣 logo"。

````python
def copy_dataset_as_class_agnostic(src_root, dst_root, prefix,
                                   negative_images=(), neg_ratio=0.15,
                                   train_only=False, max_train_images=None,
                                   exclude_cls_ids=None):
    """把多类检测集复制成单类（class 0）AABB 标签的副本。

    - train_only: 只复制 train split（外部检测集用，避免污染 in-domain 的 val 选型）
    - max_train_images: 限制 train 图片数（控训练时长）
    - exclude_cls_ids: 含这些类别的图片整张跳过（unseen-brand holdout 用）
    """
    src_root, dst_root = Path(src_root), Path(dst_root)
    st = Counter()
    splits = find_split_dirs(src_root)
    if train_only:
        splits = {k: v for k, v in splits.items() if k == "train"}

    for split_key, split_dir in splits.items():
        dst_split = "valid" if split_key == "valid" else split_key
        dst_img = ensure_dir(dst_root / dst_split / "images")
        dst_lbl = ensure_dir(dst_root / dst_split / "labels")

        for img_path in iter_images(split_dir / "images"):
            if max_train_images is not None and st["images"] >= max_train_images:
                break
            label_path = split_dir / "labels" / (img_path.stem + ".txt")
            raw_lines = []
            if label_path.exists():
                raw_lines = [l for l in label_path.read_text(
                    encoding="utf-8", errors="ignore").splitlines() if l.strip()]

            if exclude_cls_ids:
                hit = False
                for line in raw_lines:
                    parts = line.split()
                    if parts and str(parts[0]).isdigit() and int(parts[0]) in exclude_cls_ids:
                        hit = True
                        break
                if hit:
                    st["skipped_holdout_image"] += 1
                    continue

            new_stem = "%s__%s" % (prefix, img_path.stem)
            try:
                os.link(img_path, dst_img / (new_stem + img_path.suffix))
            except OSError:
                shutil.copy2(img_path, dst_img / (new_stem + img_path.suffix))
            st["images"] += 1

            out_lines = []
            for line in raw_lines:
                r = parse_yolo_line(line)
                if r is None:
                    st["bad_lines"] += 1
                    continue
                _, x1, y1, x2, y2 = r
                cx, cy = (x1 + x2) / 2, (y1 + y2) / 2
                bw, bh = x2 - x1, y2 - y1
                if bw <= 1e-4 or bh <= 1e-4:
                    st["bad_boxes"] += 1
                    continue
                out_lines.append("0 %.6f %.6f %.6f %.6f" % (cx, cy, bw, bh))
            (dst_lbl / (new_stem + ".txt")).write_text(
                "\n".join(out_lines) + ("\n" if out_lines else ""), encoding="utf-8")
            st["boxes"] += len(out_lines)

        if dst_split == "train" and negative_images:
            negs = list(negative_images)
            n_neg = int(min(len(negs), max(0, neg_ratio * st["images"])))
            for i, p in enumerate(negs[:n_neg]):
                stem = "%s__neg__%04d" % (prefix, i)
                try:
                    os.link(p, dst_img / (stem + Path(p).suffix))
                except OSError:
                    shutil.copy2(p, dst_img / (stem + Path(p).suffix))
                (dst_lbl / (stem + ".txt")).write_text("", encoding="utf-8")
                st["negatives"] += 1
    return st


if RESET_MERGED_DATA and MERGED_LOGO_DIR.exists():
    shutil.rmtree(MERGED_LOGO_DIR)
ensure_dir(MERGED_LOGO_DIR)

neg_imgs = list(iter_images(NEGATIVE_IMAGES_DIR)) if ADD_NEGATIVES else []
if ADD_NEGATIVES and not neg_imgs:
    print("ADD_NEGATIVES=True 但 %s 里没有图片，忽略" % NEGATIVE_IMAGES_DIR)

merge_stats = {}
for name, root in DOWNLOADED_LOGO_DATASETS:
    merge_stats[name] = copy_dataset_as_class_agnostic(
        root, MERGED_LOGO_DIR, prefix=name, negative_images=neg_imgs)

# ---------- 步骤2：把外部检测集（LogoDet-3K 等）也并进来 ----------
if MERGE_EXTERNAL_INTO_DETECTOR:
    for name, root in EXTERNAL_DETECTION_DATASETS:
        _excl = UNSEEN_CLS_IDS.get(name, set())
        merge_stats[name] = copy_dataset_as_class_agnostic(
            root, MERGED_LOGO_DIR, prefix=name,
            train_only=True,
            max_train_images=MERGED_MAX_EXTERNAL_IMAGES,
            exclude_cls_ids=_excl)
else:
    print("MERGE_EXTERNAL_INTO_DETECTOR=False，跳过外部检测集")

merged_splits = {k: (MERGED_LOGO_DIR / k) for k in ("train", "valid", "test")
                 if (MERGED_LOGO_DIR / k / "images").is_dir()}
assert "train" in merged_splits, "融合数据集里没有 train split"
assert "valid" in merged_splits, "融合数据集里没有 valid split，YOLO 无法训练"

merged_yaml = {"nc": 1, "names": ["logo"]}
for k, v in merged_splits.items():
    merged_yaml["train" if k == "train" else ("val" if k == "valid" else "test")] = \
        str((v / "images").resolve())
with open(MERGED_LOGO_DIR / "data.yaml", "w", encoding="utf-8") as f:
    yaml.dump(merged_yaml, f, allow_unicode=True, sort_keys=False)

print("=== 融合结果 ===")
for name, st in merge_stats.items():
    print("  %-22s images=%-6d boxes=%-7d neg=%-4d bad_lines=%-4d holdout_skip=%d"
          % (name, st["images"], st["boxes"], st["negatives"],
             st["bad_lines"], st["skipped_holdout_image"]))
print("data.yaml ->", MERGED_LOGO_DIR / "data.yaml")
````

### 4.5 合成球衣 logo 数据（步骤 3：造"没见过的品牌穿在身上"）

公开 logo 数据集是白底商品图，真实场景是"印在会褶皱、有光照、被压缩的球衣上"。
这里把 LogoDet/Logo-2341 的白底 logo **透视变形 + 颜色匹配 + 运动模糊 + JPEG 压缩**后贴到球员躯干上，
生成任意数量、任意品牌的训练样本。

这一步是解决"detector 认不出没见过 logo"最直接的手段：模型见到的是 logo 的**统计形态**，而不是 4 个特定图案。

````python
def _load_logo_rgba(path, max_side=200):
    # 读成 (rgb float32, alpha 0~1)。有透明通道用 alpha；白底图用"非白"当 mask
    try:
        im = Image.open(path).convert("RGBA")
    except Exception:
        return None
    a = np.array(im)
    rgb = a[..., :3].astype(np.float32)
    alpha = a[..., 3].astype(np.float32) / 255.0
    if float(alpha.min()) > 0.98:                      # 全不透明 -> 按白底处理
        whiteness = rgb.min(axis=2)
        mask = np.clip((250.0 - whiteness) / 250.0, 0.0, 1.0)
        if mask.mean() < 0.02:
            mask = np.ones_like(mask)
    else:
        mask = alpha
    mask = cv2.GaussianBlur(mask.astype(np.float32), (3, 3), 0)
    h, w = mask.shape[:2]
    s = max_side / float(max(h, w))
    if s < 1.0:
        nw, nh = max(1, int(w * s)), max(1, int(h * s))
        rgb = cv2.resize(rgb, (nw, nh), interpolation=cv2.INTER_AREA)
        mask = cv2.resize(mask, (nw, nh), interpolation=cv2.INTER_AREA)
    return rgb, mask


def _collect_logo_files(roots, exclude_keys, max_n=3000, per_class=4, seed=SEED):
    rng = random.Random(seed)
    files = []
    for _, root in roots:
        for leaf in find_leaf_class_dirs(root):
            if normalize_brand_name(leaf.name) in exclude_keys:
                continue
            imgs = list(iter_images(leaf))
            if imgs:
                files.extend(rng.sample(imgs, min(per_class, len(imgs))))
    rng.shuffle(files)
    return files[:max_n]


def _paste_logo(frame, rgb_logo, mask, box, rng, scale_range):
    H, W = frame.shape[:2]
    x1, y1, x2, y2 = [float(v) for v in box]
    bw, bh = x2 - x1, y2 - y1
    if bw < 20 or bh < 20:
        return None
    tw = rng.uniform(*scale_range) * bw
    hh, ww = rgb_logo.shape[:2]
    th = tw * hh / max(ww, 1.0)
    if tw < 6 or th < 6:
        return None
    tw, th = int(tw), int(th)

    logo = cv2.resize(rgb_logo, (tw, th), interpolation=cv2.INTER_AREA)
    m = cv2.resize(mask, (tw, th), interpolation=cv2.INTER_AREA)

    # 透视变形，模拟球衣曲面
    pad = 0.25
    canvas_w, canvas_h = int(tw * (1 + 2 * pad)), int(th * (1 + 2 * pad))
    src = np.float32([[0, 0], [tw, 0], [tw, th], [0, th]])
    dx, dy = tw * pad, th * pad
    def j():
        return rng.uniform(-0.10, 0.10)
    dst = np.float32([[dx + j() * tw, dy + j() * th],
                      [dx + tw + j() * tw, dy + j() * th],
                      [dx + tw + j() * tw, dy + th + j() * th],
                      [dx + j() * tw, dy + th + j() * th]])
    M = cv2.getPerspectiveTransform(src, dst)
    logo_w = cv2.warpPerspective(logo, M, (canvas_w, canvas_h), flags=cv2.INTER_LINEAR,
                                 borderMode=cv2.BORDER_CONSTANT, borderValue=(0, 0, 0))
    mask_w = cv2.warpPerspective(m, M, (canvas_w, canvas_h), flags=cv2.INTER_LINEAR,
                                 borderMode=cv2.BORDER_CONSTANT, borderValue=0.0)
    if mask_w.max() < 0.05:
        return None

    # 放到躯干区域
    tx1 = x1 + 0.10 * bw
    tx2 = max(tx1 + 1, x2 - 0.10 * bw - canvas_w)
    ty1 = y1 + 0.08 * bh
    ty2 = max(ty1 + 1, y1 + 0.60 * bh - canvas_h)
    px = int(rng.uniform(tx1, max(tx1, tx2)))
    py = int(rng.uniform(ty1, max(ty1, ty2)))
    if px < 0 or py < 0 or px + canvas_w >= W or py + canvas_h >= H:
        return None

    roi = frame[py:py + canvas_h, px:px + canvas_w].astype(np.float32)
    # 颜色匹配：把 logo 亮度拉到和局部背景接近
    bg_mean = float(np.clip(roi.mean(), 20, 235))
    fg = logo_w[mask_w > 0.5]
    if fg.size == 0:
        return None
    logo_mean = float(np.clip(fg.mean(), 20, 235))
    gain = float(np.clip(bg_mean / logo_mean, 0.55, 1.8))
    logo_w = np.clip(logo_w * gain, 0, 255)

    a = (mask_w * rng.uniform(0.85, 1.0))[..., None]
    frame[py:py + canvas_h, px:px + canvas_w] = (logo_w * a + roi * (1 - a)).astype(np.uint8)

    ys, xs = np.where(mask_w > 0.35)
    if len(xs) == 0:
        return None
    bx1, bx2 = px + xs.min(), px + xs.max()
    by1, by2 = py + ys.min(), py + ys.max()
    return (bx1, by1, bx2, by2)


def synthesize_jersey_logos(merged_dir, player_root, logo_files,
                            n_samples=6000, seed=SEED):
    rng = random.Random(seed)
    dst_img = ensure_dir(merged_dir / "train" / "images")
    dst_lbl = ensure_dir(merged_dir / "train" / "labels")
    for p in list(dst_img.glob("synth__*")):
        p.unlink()
    for p in list(dst_lbl.glob("synth__*")):
        p.unlink()

    splits = find_split_dirs(player_root)
    split_dir = splits.get("train") or next(iter(splits.values()), None)
    if split_dir is None or not logo_files:
        print("缺少球员帧或 logo 素材，跳过合成")
        return 0
    frames = list(iter_images(split_dir / "images"))[:SYNTH_MAX_SRC_IMAGES]
    if not frames:
        print("球员帧为空，跳过合成")
        return 0

    _cached = {}
    made = 0
    for i in range(n_samples):
        frame_path = frames[i % len(frames)]
        img = cv2.imread(str(frame_path))
        if img is None:
            continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        H, W = img.shape[:2]
        pboxes = [(x1 * W, y1 * H, x2 * W, y2 * H) for _, x1, y1, x2, y2 in
                  read_label_boxes(split_dir / "labels" / (frame_path.stem + ".txt"))]
        if not pboxes:
            continue

        labels = []
        for _ in range(rng.randint(1, 2)):
            lp = logo_files[rng.randrange(len(logo_files))]
            if lp not in _cached:
                _cached[lp] = _load_logo_rgba(lp)
            item = _cached[lp]
            if item is None:
                continue
            bb = _paste_logo(img, item[0], item[1], rng.choice(pboxes), rng, SYNTH_LOGO_SCALE)
            if bb:
                labels.append(bb)
        if not labels:
            continue

        # 域随机化：模糊 + JPEG 压缩 + 亮度
        if rng.random() < 0.5:
            img = cv2.GaussianBlur(img, (3, 3), rng.uniform(0.3, 1.2))
        if rng.random() < 0.6:
            img = np.clip(img.astype(np.float32) * rng.uniform(0.8, 1.2), 0, 255).astype(np.uint8)

        stem = "synth__%06d" % made
        out_path = dst_img / (stem + ".jpg")
        cv2.imwrite(str(out_path), cv2.cvtColor(img, cv2.COLOR_RGB2BGR),
                    [int(cv2.IMWRITE_JPEG_QUALITY), int(rng.uniform(35, 90))])
        lines = []
        for bx1, by1, bx2, by2 in labels:
            cx, cy = (bx1 + bx2) / 2.0 / W, (by1 + by2) / 2.0 / H
            bw, bh = (bx2 - bx1) / W, (by2 - by1) / H
            lines.append("0 %.6f %.6f %.6f %.6f" % (cx, cy, bw, bh))
        (dst_lbl / (stem + ".txt")).write_text("\n".join(lines) + "\n", encoding="utf-8")
        made += 1
    return made


if ADD_SYNTHETIC_DATA:
    _logo_roots = [(n, r) for n, r in CLASSIFICATION_ROOTS if "crops" not in n.lower()]
    if not _logo_roots:
        _logo_roots = CLASSIFICATION_ROOTS
    _logo_files = _collect_logo_files(_logo_roots, EXCLUDE_BRAND_KEYS,
                                      max_n=3000, per_class=4)
    print("可用 logo 素材:", len(_logo_files))
    n_syn = synthesize_jersey_logos(MERGED_LOGO_DIR, PLAYER_DATASET_DIR,
                                    _logo_files, n_samples=SYNTH_N_SAMPLES)
    print("生成合成球衣样本:", n_syn)
else:
    print("ADD_SYNTHETIC_DATA=False，跳过")
````

### 4.6 数据体检（训练前先看一眼）

小目标检测最怕的就是"框的平均边长只有 20 px，却按 640 训练"。这一格会把这个数打出来。

````python
def dataset_report(det_root, name, sample=300, seed=SEED):
    rng = random.Random(seed)
    det_root = Path(det_root)
    print("--- %s ---" % name)
    all_sizes = []
    for split_key, split_dir in find_split_dirs(det_root).items():
        imgs = list(iter_images(split_dir / "images"))
        n_box = 0
        for p in imgs:
            ls = split_dir / "labels" / (p.stem + ".txt")
            if ls.exists():
                n_box += sum(1 for l in ls.read_text(encoding="utf-8", errors="ignore").splitlines() if l.strip())
        print("  %-6s images=%-5d boxes=%-6d" % (split_key, len(imgs), n_box))
        for p in rng.sample(imgs, min(sample, len(imgs))):
            img = cv2.imread(str(p), cv2.IMREAD_COLOR)
            if img is None:
                continue
            h, w = img.shape[:2]
            for _, x1, y1, x2, y2 in read_label_boxes(split_dir / "labels" / (p.stem + ".txt")):
                all_sizes.append((max(1, (x2 - x1) * w), max(1, (y2 - y1) * h)))
    if not all_sizes:
        print("  (没有读到任何框)")
        return
    a = np.array(all_sizes)
    side = np.sqrt(a[:, 0] * a[:, 1])
    print("  box 边长(px) p10=%.1f p50=%.1f p90=%.1f | 宽高比 p50=%.2f"
          % (np.percentile(side, 10), np.percentile(side, 50), np.percentile(side, 90),
             np.median(a[:, 0] / a[:, 1])))
    if np.percentile(side, 50) < 40:
        print("  >> 中位边长 < 40px：建议训练/推理都用更大 imgsz（当前 %d）" % LOGO_IMGSZ)


dataset_report(MERGED_LOGO_DIR, "merged_logo_detector (class-agnostic)")

print("\n--- Doctor 分类源 ---")
grand = 0
for src_name, root in CLASSIFICATION_ROOTS:
    leaves = find_leaf_class_dirs(root)
    counts = [len(list(iter_images(d))) for d in leaves]
    if not counts:
        continue
    grand += sum(counts)
    print("  %-22s classes=%-5d images=%-7d 每类 p10=%d p50=%d p90=%d"
          % (src_name, len(leaves), sum(counts),
             np.percentile(counts, 10), np.percentile(counts, 50), np.percentile(counts, 90)))
print("  Doctor 原始样本合计:", grand)
print("  other 负类:", len(list(iter_images(OTHER_CROPS / "other"))) if (OTHER_CROPS / "other").exists() else 0)
````

## 7. 构建 Doctor 分类数据集

原则：**不按品牌关键词筛选**，把注册进来的所有叶子类别目录合并，同名品牌（大小写/空格不同）去重合并成全局 label map。
另外额外挂一个 `other` 负类（4.3 生成）。

````python
OTHER_KEY = "zzz-other-class"
OTHER_DISPLAY = "other"


def build_full_brand_index(roots):
    key_to_display, records = {}, []
    _holdout_skipped = [0]
    for source_name, root in roots:
        root = Path(root)
        if not root.exists():
            continue
        leaves = find_leaf_class_dirs(root)
        print("[scan] %-22s %5d class dirs -> %s" % (source_name, len(leaves), root))
        for leaf in leaves:
            display = leaf.name.strip()
            key = normalize_brand_name(display)
            if not key:
                continue
            if key in EXCLUDE_BRAND_KEYS:
                _holdout_skipped[0] += 1
                continue
            key_to_display.setdefault(key, display)
            records.append({"path": leaf, "class_key": key, "source": source_name})

    if _holdout_skipped[0]:
        print("[holdout] 跳过 %d 个 unseen-brand holdout 类别目录（不参与 Doctor 训练）"
              % _holdout_skipped[0])
    class_keys = sorted(key_to_display)
    class_to_idx = {k: i for i, k in enumerate(class_keys)}
    idx_to_class = {i: key_to_display[k] for k, i in class_to_idx.items()}
    return records, class_to_idx, idx_to_class


brand_records, class_to_idx, idx_to_class = build_full_brand_index(CLASSIFICATION_ROOTS)

_other_dir = OTHER_CROPS / "other"
if ADD_AUTO_OTHER_CLASS and _other_dir.exists() and any(_other_dir.iterdir()):
    class_to_idx[OTHER_KEY] = len(class_to_idx)
    idx_to_class[class_to_idx[OTHER_KEY]] = OTHER_DISPLAY
    brand_records.append({"path": _other_dir, "class_key": OTHER_KEY, "source": "AutoOther"})

num_classes = len(class_to_idx)
assert num_classes >= 2, "分类数据一个类都没有，检查 3.3 / 4.1 / 4.2"
print("\n全局分类词表: %d 类 | other 类: %s" % (num_classes, OTHER_KEY in class_to_idx))

if num_classes <= 200:
    print("!! 只有 %d 类：Kaggle 数据很可能没注册进来，回去检查 3.3 的输出" % num_classes)
````

### 7.1 Dataset / 分组划分 / DataLoader

三个关键点：

- **按源图分组划分**：v1 用样本级随机 90/10，同一帧的多个 crop 会同时落进 train 和 val，val 94% 纯属虚高；
- **单类上限** `DOCTOR_MAX_PER_CLASS`：防止某个大类把训练集吃光；
- **`1/sqrt(频率)` 加权采样**：长尾缓解，比全平衡（`1/freq`）温和，不会把小类过采样到过拟合。

````python
from sklearn.model_selection import GroupShuffleSplit

_CROP_RE = re.compile(r"^(?P<src>.+)_(?P<split>train|valid|test)_(?P<stem>.+)_(?P<box>\d+)$")


def group_key_of(path):
    stem = Path(path).stem
    m = _CROP_RE.match(stem)
    if m:
        return "crop:%s__%s" % (m.group("src"), m.group("stem"))   # 同一张原始画面的所有 crop 归一组
    return "img:" + str(Path(path).parent / stem)


def collect_samples(records, class_to_idx, max_per_class=None):
    per_class = Counter()
    samples = []
    for rec in records:
        label = class_to_idx[rec["class_key"]]
        for p in iter_images(rec["path"]):
            if max_per_class and per_class[label] >= max_per_class:
                break
            try:
                if p.stat().st_size == 0:
                    continue
            except OSError:
                continue
            samples.append((str(p), label, group_key_of(p)))
            per_class[label] += 1
    return samples, per_class


all_samples, per_class_counts = collect_samples(brand_records, class_to_idx, DOCTOR_MAX_PER_CLASS)
print("样本总数: %d | 类别数: %d" % (len(all_samples), len(per_class_counts)))
_missing = num_classes - len(per_class_counts)
if _missing:
    print("!! 有 %d 个类没有任何图片" % _missing)

groups = [g for _, _, g in all_samples]
gss = GroupShuffleSplit(n_splits=1, test_size=DOCTOR_VAL_RATIO, random_state=SEED)
tr_idx, va_idx = next(gss.split(all_samples, groups=groups))
train_samples = [all_samples[i] for i in tr_idx]
val_samples = [all_samples[i] for i in va_idx]
print("Train: %d | Val: %d | 组数: %d" % (len(train_samples), len(val_samples), len(set(groups))))

_valset_groups = {g for _, _, g in train_samples}
assert not (_valset_groups & {g for _, _, g in val_samples}), "分组划分泄漏！"


class KeepAspectPad:
    """保持长宽比缩放 + pad 成正方形：logo 大多是扁长的，直接 Resize 会拉变形。"""

    def __init__(self, size=224, fill=114, min_side=32):
        self.size, self.fill, self.min_side = size, fill, min_side

    def __call__(self, img):
        w, h = img.size
        if min(w, h) < self.min_side and w > 0 and h > 0:
            s = self.min_side / min(w, h)
            img = img.resize((max(1, int(w * s)), max(1, int(h * s))), Image.LANCZOS)
            w, h = img.size
        side = max(w, h)
        canvas = Image.new("RGB", (side, side), (self.fill, self.fill, self.fill))
        canvas.paste(img, ((side - w) // 2, (side - h) // 2))
        return canvas.resize((self.size, self.size), Image.BILINEAR)


train_tf = T.Compose([
    T.RandomApply([T.RandomAffine(degrees=8, translate=(0.06, 0.06),
                                  scale=(0.85, 1.15), shear=4)], p=0.6),
    T.RandomHorizontalFlip(),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
    T.RandomGrayscale(p=0.1),
    T.RandomApply([T.GaussianBlur(3, sigma=(0.1, 1.2))], p=0.3),   # 模拟转播压缩糊
    KeepAspectPad(224),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    T.RandomErasing(p=0.25, scale=(0.02, 0.12)),
])

val_tf = T.Compose([
    KeepAspectPad(224),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])


class BrandDataset(Dataset):
    def __init__(self, samples, transform):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label, _ = self.samples[idx]
        try:
            img = Image.open(path).convert("RGB")
        except Exception:
            img = Image.new("RGB", (224, 224), (0, 0, 0))
        return self.transform(img), label


train_set = BrandDataset(train_samples, train_tf)
val_set = BrandDataset(val_samples, val_tf)

_train_freq = Counter(l for _, l, _ in train_samples)
_w = [1.0 / math.sqrt(_train_freq[l]) for _, l, _ in train_samples]
train_sampler = WeightedRandomSampler(_w, num_samples=len(train_samples), replacement=True)

train_loader = DataLoader(train_set, batch_size=DOCTOR_BATCH_SIZE, sampler=train_sampler,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          drop_last=len(train_set) >= DOCTOR_BATCH_SIZE,
                          persistent_workers=NUM_WORKERS > 0, prefetch_factor=4 if NUM_WORKERS else None)
val_loader = DataLoader(val_set, batch_size=DOCTOR_BATCH_SIZE, shuffle=False,
                        num_workers=NUM_WORKERS, pin_memory=True,
                        persistent_workers=NUM_WORKERS > 0, prefetch_factor=4 if NUM_WORKERS else None)

print("loaders ready. train batches=%d val batches=%d" % (len(train_loader), len(val_loader)))
print("最常见的 10 类:", [(idx_to_class[i], c) for i, c in _train_freq.most_common(10)])
````

## 8. 训练 Doctor（ResNet-50）

相比 v1 的 5 epoch + 固定 lr：

- AMP 混合精度 + cosine warmup
- label smoothing 0.1
- top-1 / top-5 / macro-F1 三个指标（长尾任务上 top-1 会骗人）
- 最后在 val 上拟合**温度标定**，让置信度可用来做 unknown 判定

````python
from sklearn.metrics import f1_score


def build_scheduler(optimizer, total_steps, warmup_frac=0.05):
    warm = max(1, int(total_steps * warmup_frac))

    def lr_lambda(step):
        if step < warm:
            return step / warm
        p = (step - warm) / max(1, total_steps - warm)
        return 0.5 * (1.0 + math.cos(math.pi * min(1.0, p)))

    return torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)


def train_one_epoch(model, loader, criterion, optimizer, scheduler, scaler, epoch, total_epochs):
    model.train()
    loss_sum, correct, correct5, total = 0.0, 0, 0, 0
    t0 = time.time()
    for step, (images, labels) in enumerate(loader, 1):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)
        with autocast_ctx(USE_AMP):
            logits = model(images)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
        scaler.step(optimizer)
        scaler.update()
        scheduler.step()

        loss_sum += loss.item() * images.size(0)
        _, top5 = logits.topk(min(5, logits.size(1)), dim=1)
        correct5 += (top5 == labels.unsqueeze(1)).any(dim=1).sum().item()
        correct += (logits.argmax(1) == labels).sum().item()
        total += labels.size(0)

        if step % 200 == 0:
            print("  ep %02d/%02d  it %4d/%4d  loss %.4f  acc %.2f%%  %.0f img/s"
                  % (epoch, total_epochs, step, len(loader), loss_sum / total,
                     100 * correct / total, total / (time.time() - t0)))
    return loss_sum / max(total, 1), 100 * correct / total, 100 * correct5 / total


@torch.no_grad()
def evaluate(model, loader, criterion, collect_logits=False):
    model.eval()
    loss_sum, correct, correct5, total = 0.0, 0, 0, 0
    all_pred, all_lab, all_logits = [], [], []
    for images, labels in loader:
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        with autocast_ctx(USE_AMP):
            logits = model(images)
            loss = criterion(logits, labels)
        loss_sum += loss.item() * images.size(0)
        _, top5 = logits.topk(min(5, logits.size(1)), dim=1)
        correct5 += (top5 == labels.unsqueeze(1)).any(dim=1).sum().item()
        pred = logits.argmax(1)
        correct += (pred == labels).sum().item()
        total += labels.size(0)
        all_pred.append(pred.cpu())
        all_lab.append(labels.cpu())
        if collect_logits:
            all_logits.append(logits.float().cpu())
    res = {"loss": loss_sum / max(total, 1),
           "top1": 100 * correct / max(total, 1),
           "top5": 100 * correct5 / max(total, 1)}
    if all_lab:
        y, p = torch.cat(all_lab).numpy(), torch.cat(all_pred).numpy()
        res["macro_f1"] = 100 * f1_score(y, p, average="macro", zero_division=0)
    if collect_logits and all_logits:
        res["logits"] = torch.cat(all_logits)
        res["labels"] = torch.cat(all_lab)
    return res


def fit_temperature(logits, labels, max_iter=100):
    """在验证集 logits 上拟合温度 T，使 softmax(logits/T) 的 NLL 最小。"""
    logits = logits.float()
    nll = nn.CrossEntropyLoss()
    log_t = torch.zeros(1, requires_grad=True)
    opt = torch.optim.LBFGS([log_t], lr=0.05, max_iter=max_iter)

    def closure():
        opt.zero_grad()
        loss = nll(logits / torch.exp(log_t), labels)
        loss.backward()
        return loss

    opt.step(closure)
    return float(torch.exp(log_t).item())


print("doctor train utils ready.")
````

````python
set_seed(SEED)

model_doctor = resnet50(weights=ResNet50_Weights.DEFAULT)
model_doctor.fc = nn.Linear(model_doctor.fc.in_features, num_classes)
model_doctor = model_doctor.to(DEVICE)

criterion_train = nn.CrossEntropyLoss(label_smoothing=DOCTOR_LABEL_SMOOTH)
criterion_eval = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model_doctor.parameters(), lr=DOCTOR_LR,
                              weight_decay=DOCTOR_WEIGHT_DECAY)
scheduler = build_scheduler(optimizer, DOCTOR_EPOCHS * max(1, len(train_loader)))
scaler = make_grad_scaler(USE_AMP)

best_f1, best_state = -1.0, None
for epoch in range(1, DOCTOR_EPOCHS + 1):
    tr_loss, tr_top1, tr_top5 = train_one_epoch(
        model_doctor, train_loader, criterion_train, optimizer, scheduler, scaler,
        epoch, DOCTOR_EPOCHS)
    va = evaluate(model_doctor, val_loader, criterion_eval)
    print("ep %02d | train loss %.4f top1 %.2f%% top5 %.2f%% | val loss %.4f top1 %.2f%% top5 %.2f%% macroF1 %.2f%%"
          % (epoch, tr_loss, tr_top1, tr_top5, va["loss"], va["top1"], va["top5"], va["macro_f1"]))

    if va["macro_f1"] > best_f1:
        best_f1 = va["macro_f1"]
        best_state = {k: v.detach().cpu().clone() for k, v in model_doctor.state_dict().items()}
        print("  -> new best (macroF1 %.2f%%)" % best_f1)

assert best_state is not None
model_doctor.load_state_dict(best_state)
model_doctor.eval()
print("\nbest val macroF1: %.2f%%" % best_f1)
````

````python
# ---- 温度标定：用 val 的一个子集拟合，避免 30k x 2342 的 logits 把内存吃光 ----
val_eval = evaluate(model_doctor, val_loader, criterion_eval)

_sub_n = min(8000, len(val_set))
_sub_loader = DataLoader(torch.utils.data.Subset(val_set, list(range(_sub_n))),
                         batch_size=DOCTOR_BATCH_SIZE, shuffle=False,
                         num_workers=NUM_WORKERS, pin_memory=True)
val_sub = evaluate(model_doctor, _sub_loader, criterion_eval, collect_logits=True)
TEMPERATURE = fit_temperature(val_sub["logits"], val_sub["labels"])
print("fitted temperature T = %.4f" % TEMPERATURE)

conf = torch.softmax(val_sub["logits"] / TEMPERATURE, dim=1).max(dim=1).values
print("val top1=%.2f%% top5=%.2f%% macroF1=%.2f%% | 平均 top1 置信度(标定后)=%.3f"
      % (val_eval["top1"], val_eval["top5"], val_eval["macro_f1"], float(conf.mean())))

torch.save({
    "model_state": model_doctor.state_dict(),
    "class_to_idx": class_to_idx,
    "idx_to_class": {int(k): v for k, v in idx_to_class.items()},
    "num_classes": num_classes,
    "temperature": TEMPERATURE,
    "other_key": OTHER_KEY if OTHER_KEY in class_to_idx else None,
    "val_metrics": {k: v for k, v in val_eval.items() if k not in ("logits", "labels")},
}, DOCTOR_WEIGHTS)
print("Doctor saved ->", DOCTOR_WEIGHTS)
````

## 9.1 Doctor 的 unknown 处理（闭集 → 开放集）

Doctor 是闭集分类器：遇到训练里没有的品牌，它只能在已知类里挑一个最近的，还常常给出很高置信度。
v2 只是用一个阈值硬切 `unknown`，这只能"拒绝"，不能"识别"。

这一节把 unknown 做成一条**升级链**，逐级尝试更高成本但更 open-set 的手段：

| 级别 | 手段 | 能力 | 成本 |
|---|---|---|---|
| L1 | Doctor 自身置信度 + other 类 | 拒绝明显不是品牌的 | 0 |
| L2 | **embedding gallery**（Doctor 倒数第二层特征 → 每类原型） | 修正"见过但分类头走偏"的样本 | 0（一次前向） |
| L3 | **CLIP zero-shot**（品牌名当文本 prompt） | 真正 open-set：没见过的新品牌也能靠语义匹配识别 | 中 |
| L4 | **GroundingDINO 复核**（开放词汇检测） | 确认"这块区域到底是不是 logo"，判不了的进人工池 | 高 |

L4 的关键不是让 GDINO 说出品牌名（它做不到），而是：**当 Doctor 和 CLIP 都不确定时，判断这块区域是不是一个 logo-like 物体**。
如果是 → 记为 `new_brand_candidate` 存进 unknown pool，用于主动学习/扩充 gallery，而不是直接丢掉。

````python
import torch.nn.functional as F

_FEAT_NET = None


def _feat_net():
    # ResNet50 去掉 fc，输出 2048 维 embedding
    global _FEAT_NET
    if _FEAT_NET is None:
        _FEAT_NET = nn.Sequential(*list(model_doctor.children())[:-1]).to(DEVICE).eval()
    return _FEAT_NET


@torch.no_grad()
def doctor_embed(crops_rgb, batch_size=64):
    feats = []
    net = _feat_net()
    for i in range(0, len(crops_rgb), batch_size):
        chunk = crops_rgb[i:i + batch_size]
        x = torch.stack([val_tf(Image.fromarray(c)) for c in chunk]).to(DEVICE)
        with autocast_ctx(USE_AMP):
            f = net(x)
        feats.append(F.normalize(f.float().flatten(1), dim=1).cpu())
    return torch.cat(feats) if feats else torch.zeros(0, 2048)


def build_brand_gallery(samples, per_class=8, seed=SEED):
    """每个品牌取若干样本，embedding 求平均得到原型，再 L2 归一化。"""
    rng = random.Random(seed)
    by_cls = defaultdict(list)
    for p, l, _ in samples:
        by_cls[l].append(p)

    embs, labels = [], []
    for cls, paths in sorted(by_cls.items()):
        if OTHER_IDX is not None and cls == OTHER_IDX:
            continue
        pick = paths if len(paths) <= per_class else rng.sample(paths, per_class)
        crops = []
        for p in pick:
            try:
                crops.append(np.array(Image.open(p).convert("RGB")))
            except Exception:
                continue
        if not crops:
            continue
        e = F.normalize(doctor_embed(crops).mean(dim=0), dim=0)
        embs.append(e)
        labels.append(cls)
    G = torch.stack(embs) if embs else torch.zeros(0, 2048)
    print("gallery: %d 个品牌原型, shape=%s" % (len(labels), tuple(G.shape)))
    return G, labels


GALLERY, GALLERY_LABELS = build_brand_gallery(train_samples, per_class=GALLERY_PER_CLASS)


@torch.no_grad()
def gallery_predict(crops_rgb):
    if GALLERY.numel() == 0:
        return [(None, 0.0, 1.0)] * len(crops_rgb)
    e = doctor_embed(crops_rgb)
    sims = e @ GALLERY.T
    best = sims.argmax(dim=1)
    out = []
    for i in range(len(crops_rgb)):
        b = int(best[i])
        s = float(sims[i, b])
        out.append((idx_to_class[int(GALLERY_LABELS[b])], s, 1.0 - s))
    return out
````

````python
CLIP_TEMPLATES = ["a photo of the {} logo",
                  "a {} sponsor logo on a soccer jersey",
                  "the brand {}"]
CLIP_NEGATIVES = ["a soccer jersey", "grass on a football pitch",
                  "a crowd in a stadium", "an advertising board",
                  "a player's jersey number"]

_CLIP = {"model": None, "proc": None, "text": None, "owner": None}


def _load_clip():
    if _CLIP["model"] is not None:
        return True
    try:
        from transformers import CLIPModel, CLIPProcessor
    except Exception as e:
        print("[CLIP] transformers 不可用:", e)
        return False
    try:
        _CLIP["proc"] = CLIPProcessor.from_pretrained(CLIP_MODEL_NAME)
        _CLIP["model"] = CLIPModel.from_pretrained(CLIP_MODEL_NAME).to(DEVICE).eval()
        print("[CLIP] loaded:", CLIP_MODEL_NAME)
    except Exception as e:
        print("[CLIP] 加载失败:", e)
        return False
    return True


@torch.no_grad()
def _clip_text_bank():
    if _CLIP["text"] is not None:
        return
    names = [idx_to_class[i] for i in sorted(idx_to_class) if i != OTHER_IDX]
    texts, owner = [], []
    for tpl in CLIP_TEMPLATES:
        for n in names:
            texts.append(tpl.format(n))
            owner.append(n)
    for neg in CLIP_NEGATIVES:
        texts.append(neg)
        owner.append(None)

    embs = []
    for i in range(0, len(texts), 256):
        inp = _CLIP["proc"](text=texts[i:i + 256], return_tensors="pt",
                            padding=True, truncation=True).to(DEVICE)
        e = _CLIP["model"].get_text_features(**inp)
        embs.append(F.normalize(e.float(), dim=1).cpu())
    _CLIP["text"] = torch.cat(embs)
    _CLIP["owner"] = owner
    print("[CLIP] text bank: %d prompts (%d 品牌)" % (len(texts), len(names)))


@torch.no_grad()
def clip_predict(crops_rgb, batch_size=32):
    """零样本品牌判定：返回 [(brand, score, ok)]。"""
    if not _load_clip():
        return [(None, 0.0, False)] * len(crops_rgb)
    _clip_text_bank()
    T = _CLIP["text"]
    owner = _CLIP["owner"]
    neg_idx = [i for i, o in enumerate(owner) if o is None]
    brand_idx = defaultdict(list)
    for i, o in enumerate(owner):
        if o is not None:
            brand_idx[o].append(i)
    brand_names = list(brand_idx.keys())

    out = []
    for i in range(0, len(crops_rgb), batch_size):
        chunk = crops_rgb[i:i + batch_size]
        inp = _CLIP["proc"](images=chunk, return_tensors="pt").to(DEVICE)
        e = F.normalize(_CLIP["model"].get_image_features(**inp).float(), dim=1).cpu()
        sims = e @ T.T
        for r in range(sims.shape[0]):
            row = sims[r]
            per_brand = torch.full((len(brand_names),), -1.0)
            for bi, b in enumerate(brand_names):
                per_brand[bi] = row[brand_idx[b]].max()
            bi = int(per_brand.argmax())
            best_sim = float(per_brand[bi])
            neg_sim = float(row[neg_idx].max()) if neg_idx else 0.0
            ok = (best_sim >= CLIP_MIN_SIM) and (best_sim - neg_sim >= 0.02)
            out.append((brand_names[bi], best_sim, ok))
    return out
````

````python
_GD = {"model": None, "proc": None}


def _load_gdino():
    if _GD["model"] is not None:
        return True
    try:
        from transformers import AutoProcessor, AutoModelForZeroShotObjectDetection
    except Exception as e:
        print("[GDINO] transformers 版本不支持 GroundingDINO:", e)
        return False
    try:
        _GD["proc"] = AutoProcessor.from_pretrained(GDINO_MODEL_NAME)
        _GD["model"] = AutoModelForZeroShotObjectDetection.from_pretrained(
            GDINO_MODEL_NAME).to(DEVICE).eval()
        print("[GDINO] loaded:", GDINO_MODEL_NAME)
    except Exception as e:
        print("[GDINO] 加载失败:", e)
        return False
    return True


@torch.no_grad()
def gdino_detect(image_rgb, prompt=GDINO_PROMPT,
                 box_thr=GDINO_BOX_THR, text_thr=GDINO_TEXT_THR):
    """开放词汇检测：返回 [(xyxy, score, label)]。用来判断"这块区域是不是 logo-like 物体"。"""
    if not _load_gdino():
        return []
    try:
        inputs = _GD["proc"](images=image_rgb, text=prompt, return_tensors="pt").to(DEVICE)
        outputs = _GD["model"](**inputs)
        res = _GD["proc"].post_process_grounded_object_detection(
            outputs, inputs["input_ids"],
            box_threshold=box_thr, text_threshold=text_thr,
            target_sizes=[image_rgb.shape[:2]])[0]
    except Exception as e:
        print("[GDINO] 推理失败:", e)
        return []
    boxes = res.get("boxes", [])
    scores = res.get("scores", [])
    labels = res.get("labels", [])
    out = []
    for b, s, l in zip(boxes, scores, labels):
        out.append((np.asarray(b.tolist(), dtype=np.float32), float(s), str(l)))
    return out
````

````python
UNKNOWN_POOL_DIR = ensure_dir(ARTIFACTS_ROOT / "unknown_pool")
_unknown_counter = [0]


def _save_unknown(crop_rgb, meta):
    if not SAVE_UNKNOWN_POOL:
        return None
    _unknown_counter[0] += 1
    stem = "unk_%06d" % _unknown_counter[0]
    cv2.imwrite(str(UNKNOWN_POOL_DIR / (stem + ".jpg")),
                cv2.cvtColor(crop_rgb, cv2.COLOR_RGB2BGR))
    with open(UNKNOWN_POOL_DIR / "unknown_log.jsonl", "a", encoding="utf-8") as f:
        f.write(json.dumps(dict(meta or {}, file=stem + ".jpg"), ensure_ascii=False) + "\n")
    return stem


def escalate_unknown(crop_rgb, doc, meta=None):
    """Doctor 判 unknown 后逐级升级：gallery -> CLIP -> GroundingDINO 复核。
    返回 dict(brand, conf, source, accepted, gdino_hit)。"""
    policy = (UNKNOWN_POLICY or "none").lower()
    res = {"brand": "unknown", "conf": float(doc.get("conf", 0.0)),
           "source": "doctor", "accepted": False, "gdino_hit": False}

    # L2: 自己 embedding 的品牌原型（修"见过但分类头走偏"）
    if "gallery" in policy and "GALLERY" in globals() and GALLERY.numel() > 0:
        b, s, d = gallery_predict([crop_rgb])[0]
        if b is not None and d <= GALLERY_DIST_THR:
            res.update(brand=b, conf=s, source="gallery", accepted=True)
            return res

    # L3: CLIP zero-shot（真正的 open-set，能认没见过的品牌名）
    if "clip" in policy:
        b, s, ok = clip_predict([crop_rgb])[0]
        if ok:
            res.update(brand=b, conf=s, source="clip", accepted=True)
            return res

    # L4: GroundingDINO 复核「这块区域是不是 logo-like 物体」
    if "gdino" in policy:
        boxes = gdino_detect(crop_rgb)
        if boxes:
            box, score, label = max(boxes, key=lambda t: t[1])
            area = max(1.0, (box[2] - box[0]) * (box[3] - box[1]))
            ratio = area / float(crop_rgb.shape[0] * crop_rgb.shape[1] + 1e-6)
            if score >= 0.30 and ratio >= 0.15:
                res.update(gdino_hit=True, source="gdino",
                           brand="new_brand_candidate", accepted=False)
                _save_unknown(crop_rgb, dict(meta or {}, gdino_score=score,
                                             gdino_label=label, level="gdino"))
                return res

    _save_unknown(crop_rgb, dict(meta or {}, doctor_top5=doc.get("top5"), level="unknown"))
    return res


print("unknown 升级链 ready | policy =", UNKNOWN_POLICY)
print("可用级别: gallery=%s  clip=%s  gdino=%s"
      % ("GALLERY" in globals() and GALLERY.numel() > 0,
         "clip" in (UNKNOWN_POLICY or "").lower(),
         "gdino" in (UNKNOWN_POLICY or "").lower()))
````

````python
@torch.no_grad()
def doctor_predict(image_rgb, poly, tta=DOCTOR_USE_TTA):
    h, w = image_rgb.shape[:2]
    x1, y1, x2, y2 = poly_with_context(poly, w, h)
    if x2 - x1 < 4 or y2 - y1 < 4:
        return None
    crop = image_rgb[y1:y2, x1:x2]
    img = Image.fromarray(crop)

    batch = [val_tf(img)]
    if tta:
        batch.append(val_tf(img.transpose(Image.FLIP_LEFT_RIGHT)))
    batch = torch.stack(batch).to(DEVICE)

    with autocast_ctx(USE_AMP):
        logits = model_doctor(batch)
    probs = torch.softmax(logits.float() / TEMPERATURE, dim=1).mean(dim=0)  # TTA 在概率空间平均
    top = torch.topk(probs, k=min(5, probs.numel()))

    top1_idx = int(top.indices[0])
    top1 = float(top.values[0])
    margin = float(top.values[0] - top.values[1]) if top.values.numel() > 1 else 1.0
    is_other = OTHER_IDX is not None and top1_idx == OTHER_IDX
    is_unknown = is_other or top1 < DOCTOR_MIN_CONF or margin < DOCTOR_MIN_MARGIN

    return {
        "brand": idx_to_class[top1_idx],
        "conf": top1,
        "margin": margin,
        "is_unknown": bool(is_unknown),
        "top5": [(idx_to_class[int(i)], float(v)) for i, v in zip(top.indices, top.values)],
        "crop_xyxy": (x1, y1, x2, y2),
    }


def predict_image(image_bgr, draw=False):
    # 注意：Ultralytics 的 numpy 输入按 BGR 解释，Doctor 的 crop 按 RGB 预处理
    bgr = image_bgr
    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    h, w = rgb.shape[:2]

    player_result = player_detector(bgr, classes=PLAYER_CLASS_IDS, conf=PLAYER_CONF,
                                    imgsz=PLAYER_IMGSZ, verbose=False, device=YOLO_DEVICE)[0]
    logo_result = logo_detector(bgr, conf=LOGO_CONF, iou=LOGO_IOU, max_det=LOGO_MAX_DET,
                                imgsz=LOGO_IMGSZ, agnostic_nms=True,
                                verbose=False, device=YOLO_DEVICE)[0]

    player_polys, _ = result_polygons(player_result)
    logo_polys, logo_confs = result_polygons(logo_result)

    records = []
    for logo_poly, det_conf in zip(logo_polys, logo_confs):
        ok, contain, center_in = logo_player_match(logo_poly, player_polys)
        if REQUIRE_PLAYER_CONTAINMENT and not ok:
            continue
        pred = doctor_predict(rgb, logo_poly)
        if pred is None:
            continue
        records.append({
            "brand": "unknown" if pred["is_unknown"] else pred["brand"],
            "doctor_conf": pred["conf"],
            "margin": pred["margin"],
            "detector_conf": float(det_conf),
            "containment": contain,
            "center_in": center_in,
            "top5": pred["top5"],
            "polygon": logo_poly.tolist(),
            "xyxy": [int(v) for v in poly_to_xyxy(logo_poly, w, h)],
        })

    annotated = rgb.copy()
    if draw:
        for p in player_polys:
            cv2.polylines(annotated, [p.astype(np.int32)], True, (0, 120, 255), 1)
        for r in records:
            x1, y1, x2, y2 = r["xyxy"]
            color = (0, 200, 0) if r["brand"] != "unknown" else (255, 160, 0)
            cv2.rectangle(annotated, (x1, y1), (x2, y2), color, 2)
            cv2.putText(annotated, "%s %.2f" % (r["brand"], r["doctor_conf"]),
                        (x1, max(y1 - 6, 12)), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 1)
    return annotated, records


def show_result(image_path):
    bgr = cv2.imread(str(image_path))
    if bgr is None:
        raise FileNotFoundError(image_path)
    annotated, records = predict_image(bgr, draw=True)
    print("检出 %d 个品牌候选；前 10 条：" % len(records))
    for r in records[:10]:
        print("  %-16s doctor=%.3f margin=%.3f det=%.3f containment=%.3f"
              % (r["brand"], r["doctor_conf"], r["margin"], r["detector_conf"], r["containment"]))
    plt.figure(figsize=(14, 14))
    plt.imshow(annotated)
    plt.axis("off")
    plt.show()
    return records


# 拿一张 player 验证集的图自检
_splits = find_split_dirs(PLAYER_DATASET_DIR)
_demo = None
for _k in ("valid", "test", "train"):
    if _k in _splits:
        _imgs = list(iter_images(_splits[_k] / "images"))[:5]
        if _imgs:
            _demo = _imgs[0]
            break
if _demo is not None:
    show_result(_demo)
else:
    print("没有可用的自检图片")
````

### 9.2 多源 proposal 融合 + 几何先验（步骤 1：宽召回）

单靠一个 YOLO 检测器，"没见过的 logo"必然漏。正确结构是**宽召回 proposal → 精准 verifier**：

```
logo detector (conf=0.10)  ─┐
几何先验（每个球员框自动生成 胸口/左袖/右袖/短裤 4 个候选）─┼─> NMS 去重 ─> Doctor 批量分类 ─> unknown 升级链 ─> brand
GroundingDINO（可选，开放词汇）─┘
```

**几何先验是这里最有价值的一条**：赞助商位置高度可预测，等于给"不认识的 logo"开了一条不依赖检测器的通路。
代价是候选变多（每个球员 4 个），所以 Doctor 改成**整帧批量前向**（一次 forward 处理所有 crop）。

````python
def geometric_prior_polys(player_polys):
    """每个球员框生成 胸口 / 左袖 / 右袖 / 短裤 四个候选区。"""
    out = []
    for p in player_polys:
        x1, y1 = p.min(axis=0)
        x2, y2 = p.max(axis=0)
        w, h = max(x2 - x1, 1e-6), max(y2 - y1, 1e-6)

        def rect(fx1, fy1, fx2, fy2):
            return np.array([[x1 + fx1 * w, y1 + fy1 * h],
                             [x1 + fx2 * w, y1 + fy1 * h],
                             [x1 + fx2 * w, y1 + fy2 * h],
                             [x1 + fx1 * w, y1 + fy2 * h]], dtype=np.float32)

        out.append(rect(0.22, 0.12, 0.78, 0.42))   # chest
        out.append(rect(0.00, 0.10, 0.28, 0.38))   # left sleeve
        out.append(rect(0.72, 0.10, 1.00, 0.38))   # right sleeve
        out.append(rect(0.25, 0.52, 0.75, 0.85))   # shorts
    return out


def _poly_aabb_iou(a, b):
    ax1, ay1 = a.min(axis=0); ax2, ay2 = a.max(axis=0)
    bx1, by1 = b.min(axis=0); bx2, by2 = b.max(axis=0)
    return iou_xyxy((ax1, ay1, ax2, ay2), (bx1, by1, bx2, by2))


def nms_proposals(items, iou_thr=FUSED_NMS_IOU):
    """items: [(poly, score, source)]，按分数贪心去重。"""
    items = sorted(items, key=lambda t: -t[1])
    keep = []
    for poly, score, src in items:
        if all(_poly_aabb_iou(poly, k[0]) < iou_thr for k in keep):
            keep.append((poly, score, src))
    return keep


@torch.no_grad()
def doctor_predict_batch(image_rgb, polys, tta=DOCTOR_USE_TTA, batch_size=64):
    """整帧所有候选一次前向（含 hflip TTA）。"""
    h, w = image_rgb.shape[:2]
    crops, keep_idx = [], []
    for i, poly in enumerate(polys):
        x1, y1, x2, y2 = poly_with_context(poly, w, h)
        if x2 - x1 < 4 or y2 - y1 < 4:
            continue
        crops.append(image_rgb[y1:y2, x1:x2])
        keep_idx.append(i)
    if not crops:
        return [None] * len(polys)

    results = [None] * len(polys)
    for s in range(0, len(crops), batch_size):
        chunk = crops[s:s + batch_size]
        tensors = []
        for c in chunk:
            img = Image.fromarray(c)
            tensors.append(val_tf(img))
            if tta:
                tensors.append(val_tf(img.transpose(Image.FLIP_LEFT_RIGHT)))
        x = torch.stack(tensors).to(DEVICE)
        with autocast_ctx(USE_AMP):
            logits = model_doctor(x)
        probs = torch.softmax(logits.float() / TEMPERATURE, dim=1).reshape(len(chunk), -1, logits.size(1)).mean(dim=1)
        top = torch.topk(probs, k=min(5, probs.size(1)), dim=1)
        for r in range(len(chunk)):
            i1 = int(top.indices[r, 0])
            c1 = float(top.values[r, 0])
            margin = float(top.values[r, 0] - top.values[r, 1]) if top.values.size(1) > 1 else 1.0
            is_other = OTHER_IDX is not None and i1 == OTHER_IDX
            results[keep_idx[s + r]] = {
                "brand": idx_to_class[i1], "conf": c1, "margin": margin,
                "is_unknown": bool(is_other or c1 < DOCTOR_MIN_CONF or margin < DOCTOR_MIN_MARGIN),
                "top5": [(idx_to_class[int(i)], float(v))
                         for i, v in zip(top.indices[r], top.values[r])],
                "crop": chunk[r],
            }
    return results


def propose_and_classify(image_bgr, use_priors=USE_GEOMETRIC_PRIORS,
                         use_gdino=False, keep_unknown=False, escalate=True):
    """宽召回 proposal -> 批量分类 -> unknown 升级链。返回每条候选的完整记录。"""
    rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    h, w = rgb.shape[:2]

    player_res = player_detector(image_bgr, classes=PLAYER_CLASS_IDS, conf=PLAYER_CONF,
                                 imgsz=PLAYER_IMGSZ, verbose=False, device=YOLO_DEVICE)[0]
    logo_res = logo_detector(image_bgr, conf=LOGO_CONF, iou=LOGO_IOU, max_det=LOGO_MAX_DET,
                             imgsz=LOGO_IMGSZ, agnostic_nms=True,
                             verbose=False, device=YOLO_DEVICE)[0]
    player_polys, _ = result_polygons(player_res)
    logo_polys, logo_confs = result_polygons(logo_res)

    props = []
    for poly, conf in zip(logo_polys, logo_confs):
        ok, contain, _ = logo_player_match(poly, player_polys)
        if REQUIRE_PLAYER_CONTAINMENT and not ok:
            continue
        props.append((poly, float(conf), "yolo"))

    if use_priors:
        for poly in geometric_prior_polys(player_polys):
            props.append((poly, PRIOR_CONF, "prior"))

    if use_gdino:
        for box, score, label in gdino_detect(rgb):
            x1, y1, x2, y2 = box
            props.append((np.array([[x1, y1], [x2, y1], [x2, y2], [x1, y2]], np.float32),
                          float(score), "gdino"))

    props = nms_proposals(props)
    preds = doctor_predict_batch(rgb, [p[0] for p in props])

    records = []
    for (poly, score, source), pred in zip(props, preds):
        if pred is None:
            continue
        # 几何先验是"没有检测证据"的候选，必须用更严的置信度门槛，否则误报会爆炸
        if source == "prior" and (pred["conf"] < PRIOR_MIN_CONF or pred["margin"] < PRIOR_MIN_MARGIN):
            pred["is_unknown"] = True
        ok, contain, center_in = logo_player_match(poly, player_polys)
        rec = {
            "poly": poly,
            "xyxy": poly_to_xyxy(poly, w, h),
            "proposal_source": source,
            "proposal_score": score,
            "containment": contain,
            "accepted": (not pred["is_unknown"]),
            "brand": pred["brand"],
            "conf": pred["conf"],
            "margin": pred["margin"],
            "escalated_from": None,
        }
        if pred["is_unknown"] and escalate:
            esc = escalate_unknown(pred["crop"], pred,
                                   {"source": source, "containment": contain,
                                    "xyxy": [int(v) for v in rec["xyxy"]]})
            if esc["accepted"]:
                rec.update(accepted=True, brand=esc["brand"], conf=esc["conf"],
                           escalated_from=esc["source"])
            elif esc["brand"] == "new_brand_candidate":
                rec.update(accepted=False, brand="new_brand_candidate",
                           escalated_from="gdino")
            else:
                rec["brand"] = "unknown"
        if rec["accepted"] or keep_unknown:
            records.append(rec)
    return records
````

## 10. 分阶段评测（漏斗表）

v1 只报一个 `end-to-end acc`，看不出瓶颈在哪一格。这里把每一阶段的**条件通过率**拆开：

```
GT ──检测召回──> matched ──containment──> contained ──Doctor──> correct
      (recall)              (conditional)            (conditional)
```

另外 v1 完全没统计**误报**（FP）。对 ROI 分析来说，把广告牌误判成球衣赞助比漏检更贵，所以这里一定报 precision。

````python
IOU_THR = 0.5


def evaluate_pipeline(data_root, split="test", out_dir=None, save_bad=True,
                      max_images=None, use_priors=USE_GEOMETRIC_PRIORS,
                      oracle=True):
    data_root = Path(data_root)
    split_dir = find_split_dirs(data_root).get(split)
    if split_dir is None:
        raise FileNotFoundError("split 不存在: %s/%s" % (data_root, split))

    cfg = load_yaml(data_root / "data.yaml") if (data_root / "data.yaml").exists() else {}
    gt_names = yolo_class_names(cfg)
    img_dir, lbl_dir = split_dir / "images", split_dir / "labels"

    out_dir = Path(out_dir or (ARTIFACTS_ROOT / "eval")) / ("%s_%s" % (data_root.name, split))
    ensure_dir(out_dir)
    bad_dir = ensure_dir(out_dir / "bad_cases") if save_bad else None

    st = Counter()
    esc_src = Counter()
    brand_gt, brand_tp, brand_fp = Counter(), Counter(), Counter()
    confusion = defaultdict(Counter)

    images = list(iter_images(img_dir))
    if max_images:
        images = images[:max_images]

    for img_path in images:
        bgr = cv2.imread(str(img_path))
        if bgr is None:
            continue
        h, w = bgr.shape[:2]
        st["images"] += 1

        gts = []
        for cls_id, x1, y1, x2, y2 in read_label_boxes(lbl_dir / (img_path.stem + ".txt")):
            brand = gt_names[cls_id] if 0 <= cls_id < len(gt_names) else "class_%d" % cls_id
            gts.append({"brand": brand,
                        "xyxy": (x1 * w, y1 * h, x2 * w, y2 * h),
                        "poly": np.array([[x1 * w, y1 * h], [x2 * w, y1 * h],
                                          [x2 * w, y2 * h], [x1 * w, y2 * h]], np.float32)})
            brand_gt[brand] += 1
            st["gt"] += 1

        recs = propose_and_classify(bgr, use_priors=use_priors)
        accepted = [r for r in recs if r["accepted"]]
        st["proposals"] += len(recs)
        st["accepted"] += len(accepted)
        for r in recs:
            if r.get("escalated_from"):
                esc_src[r["escalated_from"]] += 1

        # 1) proposal 覆盖率（天花板）
        for gt in gts:
            if any(iou_xyxy(r["xyxy"], gt["xyxy"]) >= IOU_THR for r in recs):
                st["prop_hit"] += 1

        # 2) 已接受结果对 GT 的匹配（贪心，按置信度）
        used = set()
        matched_gt = {}
        for r in sorted(accepted, key=lambda x: -x["conf"]):
            best_j, best_iou = -1, IOU_THR
            for j, gt in enumerate(gts):
                if j in used:
                    continue
                iou = iou_xyxy(r["xyxy"], gt["xyxy"])
                if iou >= best_iou:
                    best_j, best_iou = j, iou
            if best_j >= 0:
                used.add(best_j)
                matched_gt[best_j] = r
            else:
                st["fp_accepted"] += 1
                brand_fp[r["brand"]] += 1
                if bad_dir is not None:
                    vis = bgr.copy()
                    px1, py1, px2, py2 = [int(v) for v in r["xyxy"]]
                    cv2.rectangle(vis, (px1, py1), (px2, py2), (0, 0, 255), 2)
                    cv2.putText(vis, "FP:%s %.2f" % (r["brand"], r["conf"]),
                                (px1, max(py1 - 6, 12)), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 1)
                    cv2.imwrite(str(bad_dir / ("%s_fp_%s.jpg" % (img_path.stem, r["brand"]))), vis)

        for j, gt in enumerate(gts):
            if oracle:
                orc = doctor_predict(cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB), gt["poly"])
                if orc is not None:
                    st["oracle_total"] += 1
                    if (not orc["is_unknown"]) and \
                       normalize_brand_name(orc["brand"]) == normalize_brand_name(gt["brand"]):
                        st["oracle_ok"] += 1

            if j not in matched_gt:
                confusion[gt["brand"]]["MISSED"] += 1
                continue
            r = matched_gt[j]
            st["matched"] += 1
            hit = normalize_brand_name(r["brand"]) == normalize_brand_name(gt["brand"])
            if hit:
                st["correct"] += 1
                brand_tp[gt["brand"]] += 1
            else:
                brand_fp[r["brand"]] += 1
                confusion[gt["brand"]][r["brand"]] += 1
                if bad_dir is not None:
                    vis = bgr.copy()
                    gx1, gy1, gx2, gy2 = [int(v) for v in gt["xyxy"]]
                    px1, py1, px2, py2 = [int(v) for v in r["xyxy"]]
                    cv2.rectangle(vis, (gx1, gy1), (gx2, gy2), (0, 255, 0), 2)
                    cv2.rectangle(vis, (px1, py1), (px2, py2), (0, 0, 255), 2)
                    cv2.putText(vis, "GT:%s" % gt["brand"], (gx1, max(gy1 - 5, 12)),
                                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)
                    cv2.putText(vis, "Pred:%s %.2f" % (r["brand"], r["conf"]),
                                (px1, max(py1 - 20, 12)), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 1)
                    cv2.imwrite(str(bad_dir / ("%s_%d.jpg" % (img_path.stem, j))), vis)

    gt_n = max(st["gt"], 1)
    res = {
        "split": split, "images": st["images"], "gt": st["gt"],
        "proposals": st["proposals"], "accepted": st["accepted"],
        "proposal_recall": st["prop_hit"] / gt_n,
        "accept_rate": st["accepted"] / max(st["proposals"], 1),
        "doctor_acc_on_matched": st["correct"] / max(st["matched"], 1),
        "doctor_oracle_acc": st["oracle_ok"] / max(st["oracle_total"], 1),
        "e2e_per_gt": st["correct"] / gt_n,
        "fp_accepted_per_image": st["fp_accepted"] / max(st["images"], 1),
        "escalation": dict(esc_src),
    }

    print("\n========== %s / %s ==========" % (data_root.name, split))
    print("images=%d  GT=%d  proposals=%d  接受=%d"
          % (res["images"], res["gt"], res["proposals"], res["accepted"]))
    print("  1) proposal 覆盖率(天花板) = %.1f%%   <-- 检测器+几何先验能框到多少"
          % (100 * res["proposal_recall"]))
    print("  2) 接受率 = %.1f%%  (unknown 升级链: %s)"
          % (100 * res["accept_rate"], res["escalation"] or "无"))
    print("  3) Doctor 在已匹配上的准确率 = %.1f%% | oracle(GT crop) = %.1f%%"
          % (100 * res["doctor_acc_on_matched"], 100 * res["doctor_oracle_acc"]))
    print("  => e2e = %.1f%%   |  误报 %.2f 个/图"
          % (100 * res["e2e_per_gt"], res["fp_accepted_per_image"]))
    print("  品牌级:")
    for b in sorted(brand_gt, key=lambda k: -brand_gt[k]):
        p = brand_tp[b] / max(brand_tp[b] + brand_fp[b], 1)
        r = brand_tp[b] / max(brand_gt[b], 1)
        f1 = 2 * p * r / max(p + r, 1e-9)
        print("    %-18s gt=%-4d recall=%.2f precision=%.2f f1=%.2f | %s"
              % (b, brand_gt[b], r, p, f1, dict(confusion[b])))

    res["per_brand"] = {b: {"gt": brand_gt[b], "tp": brand_tp[b], "fp": brand_fp[b],
                            "confusion": dict(confusion[b])} for b in brand_gt}
    (out_dir / "summary.json").write_text(json.dumps(res, indent=2, ensure_ascii=False),
                                          encoding="utf-8")
    print("saved ->", out_dir / "summary.json")
    return res
````

````python
def _unwrap_emb(out):
    if isinstance(out, torch.Tensor): return out
    if isinstance(out, (tuple, list)):
        for x in out:
            if isinstance(x, torch.Tensor) and x.dim() >= 2: return x
        return None
    if isinstance(out, dict):
        for k in ("text_embeds", "image_embeds", "pooler_output"):
            if out.get(k) is not None: return out[k]
        lhs = out.get("last_hidden_state"); return lhs[:, 0] if lhs is not None else None
    for k in ("text_embeds", "image_embeds", "pooler_output"):
        v = getattr(out, k, None)
        if v is not None: return v
    lhs = getattr(out, "last_hidden_state", None)
    return lhs[:, 0] if lhs is not None else None


def _apply_proj(emb, proj):
    if proj is None or not hasattr(proj, "weight"): return emb
    in_f, out_f = proj.weight.shape[1], proj.weight.shape[0]
    d = emb.shape[-1]
    if d == in_f and d != out_f: return proj(emb)
    if d == out_f and d != in_f: return emb
    return proj(emb)


def _clip_text_emb(model, inputs):
    tm = getattr(model, "text_model", None)
    if tm is not None:
        kw = {"input_ids": inputs["input_ids"]}
        if "attention_mask" in inputs: kw["attention_mask"] = inputs["attention_mask"]
        return _apply_proj(_unwrap_emb(tm(**kw)), getattr(model, "text_projection", None))
    return _apply_proj(_unwrap_emb(model.get_text_features(**inputs)),
                       getattr(model, "text_projection", None))


def _clip_image_emb(model, pixel_values):
    vm = getattr(model, "vision_model", None)
    if vm is not None:
        return _apply_proj(_unwrap_emb(vm(pixel_values=pixel_values)),
                           getattr(model, "visual_projection", None))
    return _apply_proj(_unwrap_emb(model.get_image_features(pixel_values=pixel_values)),
                       getattr(model, "visual_projection", None))


@torch.no_grad()
def _clip_text_bank():
    if _CLIP["text"] is not None: return
    model = _CLIP["model"]
    names = [idx_to_class[i] for i in sorted(idx_to_class) if i != OTHER_IDX]
    texts, owner = [], []
    for tpl in CLIP_TEMPLATES:
        for n in names: texts.append(tpl.format(n)); owner.append(n)
    for neg in CLIP_NEGATIVES: texts.append(neg); owner.append(None)
    embs = []
    for i in range(0, len(texts), 256):
        inp = _CLIP["proc"](text=texts[i:i+256], return_tensors="pt",
                            padding=True, truncation=True).to(DEVICE)
        embs.append(F.normalize(_clip_text_emb(model, inp).float(), dim=1).cpu())
    _CLIP["text"] = torch.cat(embs); _CLIP["owner"] = owner
    print("[CLIP] text bank: %d prompts (%d 品牌) | dim=%d"
          % (len(texts), len(names), _CLIP["text"].shape[1]))


@torch.no_grad()
def _clip_selftest():
    model = _CLIP["model"]
    inp = _CLIP["proc"](images=[np.zeros((64, 64, 3), np.uint8)], return_tensors="pt").to(DEVICE)
    e = F.normalize(_clip_image_emb(model, inp["pixel_values"]).float(), dim=1).cpu()
    if e.shape[1] != _CLIP["text"].shape[1]:
        print("!! dim 不一致 (%d vs %d)，禁用 CLIP" % (e.shape[1], _CLIP["text"].shape[1]))
        return False
    sims = e @ _CLIP["text"].T
    print("[CLIP] selftest OK | dim=%d | sim mean=%.3f max=%.3f"
          % (e.shape[1], float(sims.mean()), float(sims.max())))
    return True


@torch.no_grad()
def clip_predict(crops_rgb, batch_size=32):
    if not _load_clip(): return [(None, 0.0, False)] * len(crops_rgb)
    try:
        _clip_text_bank()
        if _CLIP.get("_verified") is None:
            _CLIP["_verified"] = bool(_clip_selftest())
        if not _CLIP["_verified"]: return [(None, 0.0, False)] * len(crops_rgb)
    except Exception as e:
        print("[CLIP] 初始化失败，降级 unknown:", e)
        return [(None, 0.0, False)] * len(crops_rgb)

    model = _CLIP["model"]
    T_bank, owner = _CLIP["text"], _CLIP["owner"]
    neg_idx = [i for i, o in enumerate(owner) if o is None]
    brand_idx = defaultdict(list)
    for i, o in enumerate(owner):
        if o is not None: brand_idx[o].append(i)
    brand_names = list(brand_idx.keys())

    out = []
    for i in range(0, len(crops_rgb), batch_size):
        chunk = crops_rgb[i:i+batch_size]
        try:
            inp = _CLIP["proc"](images=chunk, return_tensors="pt").to(DEVICE)
            e = F.normalize(_clip_image_emb(model, inp["pixel_values"]).float(), dim=1).cpu()
        except Exception as ex:
            print("[CLIP] 图像前向失败:", ex)
            out.extend([(None, 0.0, False)] * len(chunk)); continue
        sims = e @ T_bank.T
        best_per_brand = torch.stack([sims[:, brand_idx[b]].max(dim=1).values
                                      for b in brand_names], dim=1)
        bi = best_per_brand.argmax(dim=1)
        for r in range(len(chunk)):
            j = int(bi[r]); bs = float(best_per_brand[r, j])
            ns_ = float(sims[r, neg_idx].max()) if neg_idx else 0.0
            out.append((brand_names[j], bs, (bs >= CLIP_MIN_SIM) and (bs - ns_ >= 0.02)))
    return out


_CLIP["text"] = None          # 清掉旧的（错误空间的）文本库
_CLIP["_verified"] = None
````

### 10.1 unseen-brand 泛化评测（步骤 1 的度量）

上面那套是在 Sponsor-detector-2 上评的，品牌检测器全见过。这一格才回答你的问题：
**换成从没见过的品牌，logo detector 还能不能框到？**

做法：拿 LogoDet-3K 的 test split（从未进过训练），按 3.5 定义的那 150 个 holdout 品牌过滤出 GT，
只跑检测器本身（不接 Doctor、不做 containment），算框级 recall/precision。同时用同一批数据里**见过的品牌**做对照。

````python
def evaluate_detector_recall(data_root, split, class_ids=None,
                             conf=LOGO_CONF, iou_thr=0.5, max_images=None, max_gt=4000):
    """只评检测器：框级 recall / precision。class_ids=None 表示全部类别。"""
    data_root = Path(data_root)
    split_dir = find_split_dirs(data_root).get(split)
    if split_dir is None:
        raise FileNotFoundError("split 不存在: %s/%s" % (data_root, split))
    cfg = load_yaml(data_root / "data.yaml") if (data_root / "data.yaml").exists() else {}
    names = yolo_class_names(cfg)
    img_dir, lbl_dir = split_dir / "images", split_dir / "labels"

    st = Counter()
    per_brand_gt, per_brand_tp = Counter(), Counter()
    images = list(iter_images(img_dir))
    if max_images:
        images = images[:max_images]

    for img_path in images:
        bgr = cv2.imread(str(img_path))
        if bgr is None:
            continue
        h, w = bgr.shape[:2]
        gts = []
        for cls_id, x1, y1, x2, y2 in read_label_boxes(lbl_dir / (img_path.stem + ".txt")):
            brand = names[cls_id] if 0 <= cls_id < len(names) else "class_%d" % cls_id
            gts.append({"cls": cls_id, "brand": brand,
                        "xyxy": (x1 * w, y1 * h, x2 * w, y2 * h)})
        if not gts:
            continue
        sel = [g for g in gts if class_ids is None or g["cls"] in class_ids]
        if not sel:
            continue

        res = logo_detector(bgr, conf=conf, iou=LOGO_IOU, max_det=LOGO_MAX_DET,
                            imgsz=LOGO_IMGSZ, agnostic_nms=True,
                            verbose=False, device=YOLO_DEVICE)[0]
        polys, confs = result_polygons(res)
        preds = [poly_to_xyxy(p, w, h) for p in polys]

        used = set()
        matched_pred = set()
        for pi in np.argsort(-confs) if len(preds) else []:
            best_j, best_iou = -1, iou_thr
            for j, g in enumerate(gts):
                if j in used:
                    continue
                v = iou_xyxy(preds[pi], g["xyxy"])
                if v >= best_iou:
                    best_j, best_iou = j, v
            if best_j >= 0:
                used.add(best_j)
                matched_pred.add(int(pi))
                if class_ids is None or gts[best_j]["cls"] in class_ids:
                    st["tp"] += 1
                    per_brand_tp[gts[best_j]["brand"]] += 1

        st["pred"] += len(preds)
        st["pred_matched_any"] += len(matched_pred)
        for g in sel:
            st["gt"] += 1
            per_brand_gt[g["brand"]] += 1
        if st["gt"] >= max_gt:
            break

    recall = st["tp"] / max(st["gt"], 1)
    precision = st["pred_matched_any"] / max(st["pred"], 1)
    return {"gt": st["gt"], "tp": st["tp"], "pred": st["pred"],
            "recall": recall, "precision": precision,
            "per_brand": {b: {"gt": per_brand_gt[b], "tp": per_brand_tp[b]}
                          for b in per_brand_gt}}


if EXTERNAL_DETECTION_DATASETS and UNSEEN_BRAND_NAMES:
    _ds_name, _ds_root = EXTERNAL_DETECTION_DATASETS[0]
    _split = "test" if "test" in find_split_dirs(_ds_root) else "valid"
    _seen_ids = None
    _all = find_class_names_file(_ds_root) or []
    _hold = UNSEEN_CLS_IDS.get(_ds_name, set())
    _seen_ids = set(range(len(_all))) - _hold if _all else None

    print("=== %s / %s ===" % (_ds_name, _split))
    unseen = evaluate_detector_recall(_ds_root, _split, class_ids=_hold, max_images=800)
    print("recall@unseen = %.1f%%  precision=%.1f%%  (GT=%d, 覆盖 %d 个未见品牌)"
          % (100 * unseen["recall"], 100 * unseen["precision"], unseen["gt"],
             len(unseen["per_brand"])))

    if _seen_ids:
        seen = evaluate_detector_recall(_ds_root, _split, class_ids=_seen_ids, max_images=800)
        print("recall@seen   = %.1f%%  precision=%.1f%%  (GT=%d)"
              % (100 * seen["recall"], 100 * seen["precision"], seen["gt"]))
        print("\n泛化 gap = %.1f 个点（越小越好）" % (100 * (seen["recall"] - unseen["recall"])))

    (ARTIFACTS_ROOT / "eval").mkdir(parents=True, exist_ok=True)
    (ARTIFACTS_ROOT / "eval" / "unseen_brand_eval.json").write_text(
        json.dumps({"unseen": unseen}, indent=2, ensure_ascii=False), encoding="utf-8")
    print("saved ->", ARTIFACTS_ROOT / "eval" / "unseen_brand_eval.json")
else:
    print("没有外部检测集或 holdout 未定义，跳过 unseen-brand 评测")
````

## 11. 视频 / 时效聚合

真实的 ROI 分析输入是整场比赛，输出是"某品牌曝光多少秒"。单帧准确率再高，抖一帧就漏一帧也没用。

这里用 ByteTrack 给每个 logo 一个 `track_id`，把同一 track 的多帧 Doctor 概率投票聚合：

- 单帧误判会被多帧投票压掉（最便宜的 +5~8 个点，纯推理不重训）；
- 顺便得到 start / end / duration，直接产出曝光时长 CSV。

````python
def run_video(video_path, out_csv=None, max_frames=None, frame_stride=1,
              min_track_frames=3, draw_out=None):
    cap = cv2.VideoCapture(str(video_path))
    if not cap.isOpened():
        raise FileNotFoundError("打不开视频: %s" % video_path)
    fps = cap.get(cv2.CAP_PROP_FPS) or 25.0
    tracks = defaultdict(lambda: {"votes": Counter(), "probs": [], "frames": [],
                                  "t0": None, "t1": None, "boxes": []})
    writer, idx, kept = None, 0, 0

    while True:
        ok, frame = cap.read()
        if not ok:
            break
        idx += 1
        if idx % frame_stride:
            continue
        if max_frames and idx > max_frames:
            break

        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        h, w = rgb.shape[:2]

        player_res = player_detector(frame, classes=PLAYER_CLASS_IDS, conf=PLAYER_CONF,
                                     imgsz=PLAYER_IMGSZ, verbose=False, device=YOLO_DEVICE)[0]
        logo_res = logo_detector.track(frame, persist=True, conf=LOGO_CONF, iou=LOGO_IOU,
                                       max_det=LOGO_MAX_DET, imgsz=LOGO_IMGSZ,
                                       agnostic_nms=True, tracker="bytetrack.yaml",
                                       verbose=False, device=YOLO_DEVICE)[0]

        player_polys, _ = result_polygons(player_res)
        logo_polys, logo_confs = result_polygons(logo_res)
        track_ids = None
        if getattr(logo_res, "boxes", None) is not None and logo_res.boxes.id is not None:
            track_ids = logo_res.boxes.id.int().cpu().tolist()

        for k, logo_poly in enumerate(logo_polys):
            ok_c, contain, _ = logo_player_match(logo_poly, player_polys)
            if REQUIRE_PLAYER_CONTAINMENT and not ok_c:
                continue
            pred = doctor_predict(rgb, logo_poly)
            if pred is None:
                continue
            tid = track_ids[k] if (track_ids and k < len(track_ids)) else ("det_%d_%d" % (idx, k))
            t = idx / fps
            tr = tracks[tid]
            tr["votes"][pred["brand"] if not pred["is_unknown"] else "unknown"] += 1
            tr["probs"].append(pred["conf"])
            tr["frames"].append(idx)
            tr["boxes"].append([int(v) for v in poly_to_xyxy(logo_poly, w, h)])
            tr["t0"] = t if tr["t0"] is None else min(tr["t0"], t)
            tr["t1"] = t if tr["t1"] is None else max(tr["t1"], t)
            kept += 1

        if draw_out is not None:
            vis = cv2.cvtColor(rgb.copy(), cv2.COLOR_RGB2BGR)
            for p in player_polys:
                cv2.polylines(vis, [p.astype(np.int32)], True, (255, 120, 0), 1)
            for k, logo_poly in enumerate(logo_polys):
                cv2.polylines(vis, [logo_poly.astype(np.int32)], True, (0, 255, 0), 2)
            if writer is None:
                writer = cv2.VideoWriter(str(draw_out),
                                         cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
            writer.write(vis)

    cap.release()
    if writer is not None:
        writer.release()

    rows = []
    for tid, tr in tracks.items():
        n = sum(tr["votes"].values())
        if n < min_track_frames:
            continue
        brand = tr["votes"].most_common(1)[0][0]
        rows.append({
            "track_id": tid, "brand": brand, "n_frames": n,
            "vote_ratio": tr["votes"][brand] / n,
            "mean_conf": float(np.mean(tr["probs"])),
            "start_s": round(tr["t0"], 2), "end_s": round(tr["t1"], 2),
            "duration_s": round(tr["t1"] - tr["t0"] + 1.0 / fps, 2),
            "top_box": tr["boxes"][len(tr["boxes"]) // 2],
        })
    df = pd.DataFrame(rows).sort_values("duration_s", ascending=False) if rows else pd.DataFrame()

    if out_csv and len(df):
        df.to_csv(out_csv, index=False)
        print("曝光时长表 ->", out_csv)

    print("处理帧数=%d | 通过的 logo 实例=%d | 生成的 track=%d" % (idx, kept, len(df)))
    if len(df):
        agg = df.groupby("brand")["duration_s"].sum().sort_values(ascending=False)
        print("\n按品牌聚合的曝光时长（秒）:")
        print(agg.head(20).to_string())
    return df


print("run_video ready.")
````

| 类别 | 内容 |
|---|---|

| 误检 | `other` 负类、几何先验更严门槛、containment 软匹配 |

| 可度量 | proposal 覆盖率 / 接受率 / Doctor 条件准确率 / e2e / FP per image / unseen-brand gap |
| 时序 | ByteTrack + track 级概率投票 → 曝光时长 CSV |

### 还不算 bug，但仍是短板

| 短板 | 为什么重要 | 怎么改 |
|---|---|---|
| **1. 单帧检测，track 间不合并** | 同一件球衣换 track ID 就断成两段，曝光时长被切碎 | 轨迹级 ReID 特征聚类合并 track；关键帧高分辨率重检 |
| **2. 检测器分不出赞助位置** | 胸口 sponsor 与袖口/短裤价值完全不同 | 用 logo 相对球员框的归一化坐标做 KMeans/小分类，划 chest / sleeve / shorts |
| **3. 不知道是哪个球员** | 曝光归因需要"球员 × 品牌 × 时长" | 球衣号 OCR（PARSeq/TrOCR）或 ReID 绑定 track |
| **4. 曝光时长没按可见面积加权** | 被遮挡的 logo 按满时长计费会高估 | `Σ visible_ratio × dt`，visible_ratio = 框∩人框 / 框 |
| **5. gallery 只存了 8 张/类的均值** | 多模态品牌（不同颜色/横版竖版）会被平均掉 | 每类存 K 个簇心（k-means），或直接存 top-k 检索而非原型 |
| **6. CLIP 词表太重** | 2300+ 品牌 × 3 prompt 的 zero-shot 噪声大 | 只对"该联赛可能出现的 50~100 个品牌"建词表，其余走 gallery + 人工池 |
| **7. GroundingDINO 只当二值复核用** | 它其实能做开放词汇框，只用来判"是不是 logo"浪费了 | 给它 prompt 品牌名列表，做零样本检测并把框并入 proposal |
| **8. 合成数据的域 gap 还在** | 贴图再像也不等于真实转播 | 用真实比赛片段做**自监督预训练**（DINO/MAE），或对合成数据做人脸/球衣分割式的 alpha 合成 |
| **9. unknown pool 没有闭环** | 现在只是存下来，没有反哺 | 每周人工审 200 张 → 加入 gallery 与检测训练集（主动学习闭环） |
| **10. 只有单 seed** | 85% 与 87% 可能只是随机波动 | 3 个 seed 报均值±方差；对 imgsz / 先验门槛 / TTA 做消融 |
| **11. 没有推理性能预算** | 一场 90 分钟 16 万帧，两阶段跑不完 | TensorRT/ONNX + 关键帧策略（track 起始帧跑 Doctor，后续复用）+ 帧率自适应跳帧 |
| **12. 没有 conformal 保证** | 下游要"误报率 ≤ α"这种承诺 | 在 val 上按品牌算阈值，做 conformal prediction |

### 建议的下一步

1. **unknown pool 闭环**（#9）—— 这是让系统"越用越准"的唯一路径，成本最低；
2. **几何先验 + 位置分类 + 球衣号**（#2 #3）—— 把报表从品牌级做到球员级；
3. **可见面积加权 + track 合并**（#1 #4）—— 让曝光时长真正可交付；
4. **TensorRT + 关键帧策略**（#11）—— 从"能跑"到"能跑完整场"。

### 常见问题

- **显卡 OOM**：`LOGO_BATCH` 降到 4，或 `LOGO_BASE_MODEL="yolov8n.pt"`、`LOGO_IMGSZ=960`。
- **训练太慢**：调小 `MERGE_EXTERNAL_INTO_DETECTOR` 的 `MERGED_MAX_EXTERNAL_IMAGES` 和 `SYNTH_N_SAMPLES`。
- **CLIP / GroundingDINO 加载失败**：把 `UNKNOWN_POLICY` 改成 `"gallery"` 或 `"none"`，其余流程不受影响。
- **`new_brand_candidate` 太多**：说明 Doctor 词表覆盖不足，去看 `artifacts/unknown_pool/`，人工确认后把品牌加进 `CUSTOM_BRAND_DIR` 重训。
- **几何先验误报多**：调高 `PRIOR_MIN_CONF` / `PRIOR_MIN_MARGIN`，或关掉 `USE_GEOMETRIC_PRIORS`。
- **Kaggle 检测集缺类别名**：Doctor 只会得到 `class_<id>`，在 `EXTERNAL_CLASS_NAMES_FILES` 里补类别名文件（同时 unseen holdout 也依赖它）。
