# 🏅 Jersey Sponsor Logo Detection & Brand Classification (Clean Pipeline)

这个 notebook 只保留 **最终有效的三段式 pipeline**，并按照“数据准备 → 检测器训练 → 分类器训练 → 端到端推理”的顺序重新组织。

## 最终架构

| 阶段 | 模型 | 作用 | 训练数据 |
|---|---|---|---|
| Player Detector | YOLOv8n | 检测球员/守门员，提供人体 OBB | `soccer-iqjx5` player dataset |
| Logo Detector | YOLOv8n (class-agnostic) | 只负责框出 logo 候选区域 | 多个 detection 数据集融合 |
| Doctor | ResNet-50 | 对 logo crop 做品牌多分类 | **完整融合分类训练集**（LogoDet-3K + 2341 类 logo 数据集 + 本地 crop + 自行爬取/整理图片），**不做关键词筛选** |

## 推理时的硬性过滤

1. Player Detector 先得到人体 OBB；
2. Logo Detector 得到 logo OBB；
3. 只有 logo OBB **完全落在某个人体 OBB 内**（默认 `containment_ratio >= 1.0`）才送去 Doctor 分类，避免背景广告牌干扰。

> 注意：原始 notebook 中的 2 类 doctor、5 类 keyword-filtered doctor、重复训练、ExposureEngine 探索、旧版 pipeline 等都已删除。当前文件按从上到下的顺序即可复现最终 pipeline。

## 2. 通用工具函数

包括数据集扫描、YOLO 标签读写、OBB/AABB containment 计算等。所有阶段共用，避免每个 block 重复定义。

````python
IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tif", ".tiff"}


def set_seed(seed: int = 42) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def ensure_dir(path) -> Path:
    path = Path(path)
    path.mkdir(parents=True, exist_ok=True)
    return path


def load_yaml(path) -> dict:
    with open(path, "r", encoding="utf-8") as f:
        return yaml.safe_load(f) or {}


def yolo_class_names(data_yaml: dict) -> list:
    names = data_yaml.get("names", [])
    if isinstance(names, dict):
        return [names[k] for k in sorted(names, key=lambda x: int(x))]
    return list(names)


def iter_images(directory):
    directory = Path(directory)
    if not directory.exists():
        return
    for path in sorted(directory.iterdir()):
        if path.is_file() and path.suffix.lower() in IMAGE_EXTS:
            yield path


def read_label_boxes(label_path):
    """读取 YOLO detect / OBB 标签，统一返回 [(cls_id, x1, y1, x2, y2)]（归一化）。"""
    label_path = Path(label_path)
    if not label_path.exists():
        return []
    boxes = []
    for line in label_path.read_text(encoding="utf-8", errors="ignore").splitlines():
        parts = line.split()
        if len(parts) < 5:
            continue
        cls_id = int(float(parts[0]))
        coords = [float(v) for v in parts[1:]]
        if len(coords) == 4:  # x_center, y_center, w, h
            cx, cy, bw, bh = coords
            boxes.append((cls_id, cx - bw / 2, cy - bh / 2, cx + bw / 2, cy + bh / 2))
        elif len(coords) >= 8:  # OBB: x1,y1,x2,y2,x3,y3,x4,y4
            xs, ys = coords[0::2], coords[1::2]
            boxes.append((cls_id, min(xs), min(ys), max(xs), max(ys)))
    return boxes


def normalize_brand_name(name: str) -> str:
    """用归一化 key 合并不同来源里同名但大小写/空格不同的品牌。"""
    return re.sub(r"[^a-z0-9]+", "", str(name).lower())


def sanitize_name(name: str) -> str:
    name = re.sub(r"[^A-Za-z0-9._ -]+", "_", str(name)).strip()
    return name or "unknown"


def find_leaf_class_dirs(root) -> list:
    """递归找出“直接包含图片”的目录，作为 ImageFolder 的类别目录。"""
    root = Path(root)
    if not root.exists():
        return []
    result = []
    directories = [root] + [p for p in root.rglob("*") if p.is_dir()]
    for directory in directories:
        try:
            if any(p.is_file() and p.suffix.lower() in IMAGE_EXTS for p in directory.iterdir()):
                result.append(directory)
        except (PermissionError, OSError):
            continue
    return sorted(set(result))


def register_classification_root(name: str, root) -> None:
    root = Path(root)
    if not root.exists():
        print(f"[skip] {name}: path not found -> {root}")
        return
    resolved = str(root.resolve())
    for _, existing in CLASSIFICATION_ROOTS:
        if str(Path(existing).resolve()) == resolved:
            return
    CLASSIFICATION_ROOTS.append((name, root))
    print(f"[registered] {name}: {root}")


def containment_ratio(inner_poly, outer_poly) -> float:
    """
    计算 inner 凸多边形落在 outer 凸多边形内部的比例（0~1）。
    OBB 的四个角点正好构成凸多边形，cv2.intersectConvexConvex 可以精确计算交集面积。
    """
    inner = np.asarray(inner_poly, dtype=np.float32).reshape(-1, 2)
    outer = np.asarray(outer_poly, dtype=np.float32).reshape(-1, 2)
    inner_area = abs(cv2.contourArea(inner))
    if inner_area <= 1e-6:
        return 0.0
    inter_area, _ = cv2.intersectConvexConvex(inner, outer)
    return float(inter_area / inner_area)


def result_polygons(result):
    """从 Ultralytics result 中同时兼容 OBB 与普通检测框，返回 polygon + confidence。"""
    obb = getattr(result, "obb", None)
    if obb is not None and len(obb) > 0:
        polys = obb.xyxyxyxy.cpu().numpy().reshape(-1, 4, 2)
        confs = obb.conf.cpu().numpy()
        return list(polys), np.asarray(confs)

    boxes = getattr(result, "boxes", None)
    if boxes is not None and len(boxes) > 0:
        polys, confs = [], []
        for box, conf in zip(boxes.xyxy.cpu().numpy(), boxes.conf.cpu().numpy()):
            x1, y1, x2, y2 = box
            polys.append(np.array([[x1, y1], [x2, y1], [x2, y2], [x1, y2]], dtype=np.float32))
            confs.append(conf)
        return polys, np.asarray(confs)

    return [], np.asarray([])


set_seed(SEED)
print("Utilities ready.")
````

## 4. 检测器数据整理

### 4.1 从原始多类 detection 标签中提取品牌 crop（供 Doctor 使用）

必须在 class-agnostic 转换 **之前** 做，否则品牌类别信息会丢失。

````python
def extract_brand_crops_from_yolo(src_root, out_root, prefix: str) -> int:
    """从原始 YOLO 多类标签中抠出 logo crop，并按品牌名归档。"""
    src_root = Path(src_root)
    out_root = ensure_dir(out_root)
    data_yaml_path = src_root / "data.yaml"
    names = yolo_class_names(load_yaml(data_yaml_path)) if data_yaml_path.exists() else []

    generic_names = {"logo", "sponsor", "brand", "0", "class_0"}
    count = 0
    for split in ("train", "valid", "test"):
        img_dir = src_root / split / "images"
        lbl_dir = src_root / split / "labels"
        if not img_dir.exists():
            continue
        for img_path in iter_images(img_dir):
            image = cv2.imread(str(img_path))
            if image is None:
                continue
            h, w = image.shape[:2]
            label_path = lbl_dir / f"{img_path.stem}.txt"
            for box_idx, (cls_id, x1n, y1n, x2n, y2n) in enumerate(read_label_boxes(label_path)):
                brand = names[cls_id] if 0 <= cls_id < len(names) else f"class_{cls_id}"
                if str(brand).strip().lower() in generic_names:
                    continue

                x1 = max(0, int(x1n * w))
                y1 = max(0, int(y1n * h))
                x2 = min(w, int(x2n * w))
                y2 = min(h, int(y2n * h))
                if x2 <= x1 or y2 <= y1:
                    continue

                brand_dir = ensure_dir(out_root / sanitize_name(brand))
                crop_name = f"{prefix}_{split}_{img_path.stem}_{box_idx}.jpg"
                cv2.imwrite(str(brand_dir / crop_name), image[y1:y2, x1:x2])
                count += 1
    return count


total_crops = 0
for name, root in LOGO_DET_SOURCES:
    n = extract_brand_crops_from_yolo(root, LOCAL_BRAND_CROPS, prefix=name)
    total_crops += n
    print(f"[crops] {name}: {n}")

if total_crops > 0:
    register_classification_root("LocalBrandCrops", LOCAL_BRAND_CROPS)
else:
    print("⚠️ 没有提取到带具体品牌名的本地 crop；如果 detection 数据集本身只有 'logo' 单类，这是正常的。")
````

### 4.2 融合多个 Logo detection 数据集，生成 class-agnostic 副本

- 不修改原始数据；
- 所有类别 id 强制改为 `0`；
- 同时兼容 YOLO detect 标签和 OBB 标签（OBB 只改 class，坐标原样保留）；
- 自动生成 `data.yaml`。

````python
def copy_dataset_as_class_agnostic(src_root, dst_root, prefix: str) -> dict:
    src_root, dst_root = Path(src_root), Path(dst_root)
    stats = {"images": 0, "labels": 0}
    for split in ("train", "valid", "test"):
        src_img = src_root / split / "images"
        if not src_img.exists():
            continue
        src_lbl = src_root / split / "labels"
        dst_img = ensure_dir(dst_root / split / "images")
        dst_lbl = ensure_dir(dst_root / split / "labels")

        for img_path in iter_images(src_img):
            new_stem = f"{prefix}__{img_path.stem}"
            shutil.copy2(img_path, dst_img / f"{new_stem}{img_path.suffix}")
            stats["images"] += 1

            label_path = src_lbl / f"{img_path.stem}.txt"
            out_lines = []
            if label_path.exists():
                for line in label_path.read_text(encoding="utf-8", errors="ignore").splitlines():
                    parts = line.split()
                    if len(parts) >= 5:
                        parts[0] = "0"
                        out_lines.append(" ".join(parts))
            (dst_lbl / f"{new_stem}.txt").write_text(
                "\n".join(out_lines) + ("\n" if out_lines else ""),
                encoding="utf-8",
            )
            stats["labels"] += len(out_lines)
    return stats


if RESET_MERGED_DATA and MERGED_LOGO_DIR.exists():
    shutil.rmtree(MERGED_LOGO_DIR)
ensure_dir(MERGED_LOGO_DIR)

merge_stats = {}
for name, root in LOGO_DET_SOURCES:
    merge_stats[name] = copy_dataset_as_class_agnostic(root, MERGED_LOGO_DIR, prefix=name)

merged_yaml = {
    "train": str((MERGED_LOGO_DIR / "train" / "images").resolve()),
    "val": str((MERGED_LOGO_DIR / "valid" / "images").resolve()),
    "test": str((MERGED_LOGO_DIR / "test" / "images").resolve()),
    "nc": 1,
    "names": ["logo"],
}
with open(MERGED_LOGO_DIR / "data.yaml", "w", encoding="utf-8") as f:
    yaml.dump(merged_yaml, f, allow_unicode=True, sort_keys=False)

print("Merged logo detector dataset:")
for k, v in merge_stats.items():
    print(f"  - {k}: {v}")
print("data.yaml ->", MERGED_LOGO_DIR / "data.yaml")
````

### 4.3 可选的训练前数据清洗：只保留落在人身上的 logo 框

最终推理阶段本来就会用 player OBB 做 containment 过滤，所以这一步 **默认关闭**。
如果你发现 logo detector 学到太多背景广告牌，可以打开 `FILTER_LOGO_TRAIN_ON_PERSON = True` 再运行这一格。

````python
def filter_logo_boxes_by_person(dataset_root, threshold=0.8, batch_size=32):
    person_model = YOLO("yolov8n.pt")
    dataset_root = Path(dataset_root)
    for split in ("train", "valid", "test"):
        img_dir = dataset_root / split / "images"
        lbl_dir = dataset_root / split / "labels"
        if not img_dir.exists():
            continue
        images = list(iter_images(img_dir))
        for i in range(0, len(images), batch_size):
            batch = images[i:i + batch_size]
            results = person_model([str(p) for p in batch], classes=[0], verbose=False, device=YOLO_DEVICE)
            for img_path, result in zip(batch, results):
                h, w = result.orig_shape
                person_boxes = result.boxes.xyxy.cpu().numpy() if result.boxes is not None else []

                label_path = lbl_dir / f"{img_path.stem}.txt"
                if not label_path.exists():
                    continue

                kept_lines = []
                for line in label_path.read_text(encoding="utf-8", errors="ignore").splitlines():
                    parts = line.split()
                    if len(parts) < 5:
                        continue
                    coords = [float(v) for v in parts[1:]]
                    if len(coords) == 4:
                        cx, cy, bw, bh = coords
                        x1n, y1n, x2n, y2n = cx - bw / 2, cy - bh / 2, cx + bw / 2, cy + bh / 2
                    else:
                        xs, ys = coords[0::2], coords[1::2]
                        x1n, y1n, x2n, y2n = min(xs), min(ys), max(xs), max(ys)

                    logo_poly = np.array([
                        [x1n * w, y1n * h], [x2n * w, y1n * h],
                        [x2n * w, y2n * h], [x1n * w, y2n * h],
                    ], dtype=np.float32)
                    if any(containment_ratio(logo_poly, p) >= threshold for p in person_boxes):
                        kept_lines.append(line)

                label_path.write_text("\n".join(kept_lines) + ("\n" if kept_lines else ""), encoding="utf-8")


if FILTER_LOGO_TRAIN_ON_PERSON:
    filter_logo_boxes_by_person(MERGED_LOGO_DIR, threshold=0.8)
    print("✅ Logo detector 训练集已完成 person-containment 清洗。")
else:
    print("跳过训练前清洗（pipeline 推理阶段会做 containment 过滤）。")
````

## 7. 构建 Doctor 的完整分类数据集

原则：**不筛选类别**。把 `CLASSIFICATION_ROOTS` 里所有数据源的叶子类别目录全部合并，同名品牌去重合并，得到全局 label map。

````python
def build_full_brand_index(root_sources):
    key_to_display = {}
    records = []

    for source_name, root in root_sources:
        root = Path(root)
        if not root.exists():
            continue
        leaves = find_leaf_class_dirs(root)
        print(f"[scan] {source_name}: {len(leaves)} class dirs -> {root}")
        for leaf in leaves:
            display_name = leaf.name.strip()
            key = normalize_brand_name(display_name)
            if not key:
                continue
            key_to_display.setdefault(key, display_name)
            records.append({"path": leaf, "class_key": key, "source": source_name})

    class_keys = sorted(key_to_display.keys())
    class_to_idx = {key: idx for idx, key in enumerate(class_keys)}
    idx_to_class = {idx: key_to_display[key] for key, idx in class_to_idx.items()}
    return records, class_to_idx, idx_to_class


# 确保本地 crop 也被纳入（如果前面提取到了具体品牌）
if LOCAL_BRAND_CROPS.exists() and any(LOCAL_BRAND_CROPS.iterdir()):
    register_classification_root("LocalBrandCrops", LOCAL_BRAND_CROPS)

brand_records, class_to_idx, idx_to_class = build_full_brand_index(CLASSIFICATION_ROOTS)
num_classes = len(class_to_idx)

if num_classes == 0:
    raise RuntimeError("没有任何分类数据被注册。请先运行 3.3 / 3.4 / 4.1。")

print(f"\n🔥 完整全局分类词表构建完成：{num_classes} 个品牌类别（未做任何关键词筛选）")
````

### 7.1 PyTorch Dataset / DataLoader

全量样本随机 90/10 划分 train/val。训练集使用数据增强，验证集只做 resize + normalize。

````python
class FullBrandDataset(Dataset):
    def __init__(self, records, class_to_idx, transform=None):
        self.transform = transform
        self.samples = []
        self.source_counts = {}

        for record in records:
            source = record["source"]
            class_idx = class_to_idx[record["class_key"]]
            n = 0
            for img_path in iter_images(record["path"]):
                self.samples.append((str(img_path), class_idx))
                n += 1
            self.source_counts[source] = self.source_counts.get(source, 0) + n

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        try:
            image = Image.open(img_path).convert("RGB")
        except Exception:
            image = Image.new("RGB", (224, 224), (0, 0, 0))
        if self.transform:
            image = self.transform(image)
        return image, label


train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

full_dataset = FullBrandDataset(brand_records, class_to_idx, transform=train_transform)
print("每个数据源贡献的样本数：")
for source, count in full_dataset.source_counts.items():
    print(f"  - {source}: {count}")
print(f"总样本数: {len(full_dataset)}")

train_size = int(0.9 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_set, val_set = random_split(
    full_dataset,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(SEED),
)

# 验证集必须使用无增强 transform；random_split 共享底层 dataset，这里用 wrapper 覆盖 transform
class TransformSubset(Dataset):
    def __init__(self, subset, transform):
        self.subset = subset
        self.transform = transform

    def __len__(self):
        return len(self.subset)

    def __getitem__(self, idx):
        img_path, label = self.subset.dataset.samples[self.subset.indices[idx]]
        image = Image.open(img_path).convert("RGB")
        return self.transform(image), label


val_set = TransformSubset(val_set, val_transform)

train_loader = DataLoader(
    train_set, batch_size=DOCTOR_BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=True,
)
val_loader = DataLoader(
    val_set, batch_size=DOCTOR_BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=True,
)

print(f"Train: {len(train_set)} | Val: {len(val_set)}")
````

## 8. 训练 Doctor（ResNet-50，全量品牌词表）

- 输出维度 = 完整融合后的品牌数；
- 保存 best checkpoint + label map；
- 不在类别层面做任何裁剪。

````python
def train_one_epoch(model, loader, criterion, optimizer, device, epoch, total_epochs):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    start = time.time()

    for batch_idx, (images, labels) in enumerate(loader):
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        outputs = model(images)
        loss = criterion(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        _, preds = torch.max(outputs, 1)
        correct += torch.sum(preds == labels).item()
        total += labels.size(0)

        if (batch_idx + 1) % 100 == 0:
            print(
                f"Epoch {epoch:02d}/{total_epochs} | Batch {batch_idx + 1}/{len(loader)} | "
                f"Loss {running_loss / total:.4f} | Acc {100 * correct / total:.2f}% | "
                f"{total / (time.time() - start):.1f} img/s"
            )

    return running_loss / total, 100 * correct / total


@torch.no_grad()
def validate_one_epoch(model, loader, criterion, device):
    model.eval()
    running_loss, correct, total = 0.0, 0, 0
    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)
        outputs = model(images)
        loss = criterion(outputs, labels)

        running_loss += loss.item() * images.size(0)
        _, preds = torch.max(outputs, 1)
        correct += torch.sum(preds == labels).item()
        total += labels.size(0)
    return running_loss / total, 100 * correct / total


model_doctor = resnet50(weights=ResNet50_Weights.DEFAULT)
model_doctor.fc = nn.Linear(model_doctor.fc.in_features, num_classes)
model_doctor = model_doctor.to(DEVICE)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(
    model_doctor.parameters(), lr=DOCTOR_LR, weight_decay=DOCTOR_WEIGHT_DECAY
)

best_val_acc = -1.0
for epoch in range(1, DOCTOR_EPOCHS + 1):
    train_loss, train_acc = train_one_epoch(
        model_doctor, train_loader, criterion, optimizer, DEVICE, epoch, DOCTOR_EPOCHS
    )
    val_loss, val_acc = validate_one_epoch(model_doctor, val_loader, criterion, DEVICE)

    print(
        f"\nEpoch {epoch:02d}/{DOCTOR_EPOCHS} | "
        f"Train Loss {train_loss:.4f} Acc {train_acc:.2f}% | "
        f"Val Loss {val_loss:.4f} Acc {val_acc:.2f}%\n"
    )

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(
            {
                "model_state": model_doctor.state_dict(),
                "class_to_idx": class_to_idx,
                "idx_to_class": idx_to_class,
                "num_classes": num_classes,
                "val_acc": val_acc,
            },
            DOCTOR_WEIGHTS,
        )
        print(f"✅ 保存新的 best doctor -> {DOCTOR_WEIGHTS} (val acc {val_acc:.2f}%)")

with open(DOCTOR_LABEL_MAP, "w", encoding="utf-8") as f:
    json.dump(
        {
            "class_to_idx": class_to_idx,
            "idx_to_class": {str(k): v for k, v in idx_to_class.items()},
            "num_classes": num_classes,
        },
        f,
        ensure_ascii=False,
        indent=2,
    )

print(f"\n🎉 Doctor 训练完成。Best val acc: {best_val_acc:.2f}%")
print("Weights:", DOCTOR_WEIGHTS)
print("Label map:", DOCTOR_LABEL_MAP)
````

## 9. 端到端推理

流程：

```
image
  ├─ Player Detector(YOLOv8n) -> player OBBs
  └─ Logo Detector(YOLOv8n)  -> logo OBBs
          └─ 对每个 logo OBB：
               若被任意 player OBB 完全包含（containment_ratio >= threshold）
               -> crop -> ResNet-50 Doctor -> brand
```

> 如果训练数据是 OBB 标注，`result_polygons()` 会自动读取 `result.obb.xyxyxyxy`；普通检测框会自动转成矩形多边形。

````python
# 载入三个模型
player_detector = YOLO(str(PLAYER_WEIGHTS))
logo_detector = YOLO(str(LOGO_WEIGHTS))

# 只把“球员/守门员”类别当作人体框；如果模型只训练了这两类，得到的就是全部类别。
PLAYER_CLASS_IDS = [
    idx for idx, name in player_detector.names.items()
    if str(name).strip().lower() in {"joueur", "gardien", "player", "goalkeeper", "keeper"}
] or None
print("Inference player class ids:", PLAYER_CLASS_IDS)

try:
    checkpoint = torch.load(str(DOCTOR_WEIGHTS), map_location=DEVICE, weights_only=False)
except TypeError:
    checkpoint = torch.load(str(DOCTOR_WEIGHTS), map_location=DEVICE)
if isinstance(checkpoint, dict) and "model_state" in checkpoint:
    state_dict = checkpoint["model_state"]
    idx_to_class = {int(k): v for k, v in checkpoint["idx_to_class"].items()}
else:
    state_dict = checkpoint
    label_map = json.loads(DOCTOR_LABEL_MAP.read_text(encoding="utf-8"))
    idx_to_class = {int(k): v for k, v in label_map["idx_to_class"].items()}

model_doctor = resnet50()
model_doctor.fc = nn.Linear(model_doctor.fc.in_features, len(idx_to_class))
model_doctor.load_state_dict(state_dict)
model_doctor = model_doctor.to(DEVICE).eval()

preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

print(f"Loaded player detector: {PLAYER_WEIGHTS}")
print(f"Loaded logo detector:   {LOGO_WEIGHTS}")
print(f"Loaded doctor:          {DOCTOR_WEIGHTS} ({len(idx_to_class)} classes)")
````

````python
def run_pipeline(image_path, show=True):
    image_bgr = cv2.imread("/content/close_ups/frame_00191.jpg")
    if image_bgr is None:
        raise FileNotFoundError(f"无法读取图片: {image_path}")
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    annotated = image_rgb.copy()
    h, w = image_rgb.shape[:2]

    player_result = player_detector(
        image_rgb, classes=PLAYER_CLASS_IDS, verbose=False, device=YOLO_DEVICE
    )[0]
    logo_result = logo_detector(image_rgb, verbose=False, device=YOLO_DEVICE)[0]

    player_polys, _ = result_polygons(player_result)
    logo_polys, logo_confs = result_polygons(logo_result)

    # 画出 player OBB（蓝色细线），方便检查 containment
    for poly in player_polys:
        cv2.polylines(annotated, [poly.astype(np.int32)], True, (0, 120, 255), 1)

    records = []
    for logo_poly, det_conf in zip(logo_polys, logo_confs):
        ratios = [containment_ratio(logo_poly, p) for p in player_polys]
        best_ratio = max(ratios) if ratios else 0.0

        if REQUIRE_PLAYER_CONTAINMENT and best_ratio < CONTAINMENT_THRESHOLD:
            continue

        x1, y1 = np.floor(logo_poly.min(axis=0)).astype(int)
        x2, y2 = np.ceil(logo_poly.max(axis=0)).astype(int)
        x1, y1 = max(0, x1), max(0, y1)
        x2, y2 = min(w, x2), min(h, y2)
        if x2 <= x1 or y2 <= y1:
            continue

        crop = image_rgb[y1:y2, x1:x2]
        tensor = preprocess(Image.fromarray(crop)).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits = model_doctor(tensor)
            prob = torch.softmax(logits, dim=1)
            pred_idx = int(torch.argmax(prob, dim=1).item())
            pred_conf = float(prob[0, pred_idx].item())

        brand = idx_to_class[pred_idx]
        label = f"{brand} {pred_conf:.2f}"
        cv2.polylines(annotated, [logo_poly.astype(np.int32)], True, (0, 255, 0), 2)
        cv2.putText(
            annotated, label,
            (x1, max(y1 - 8, 15)),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2,
        )

        records.append({
            "brand": brand,
            "doctor_conf": pred_conf,
            "detector_conf": float(det_conf),
            "containment": best_ratio,
            "polygon": logo_poly.tolist(),
        })

    print(f"Player boxes: {len(player_polys)} | Logo candidates: {len(logo_polys)} | Kept & classified: {len(records)}")
    for r in records:
        print(r)

    if show:
        plt.figure(figsize=(12, 12))
        plt.imshow(annotated)
        plt.axis("off")
        plt.title("Player Detector -> Logo Detector -> Containment -> ResNet Doctor")
        plt.show()

    return annotated, records


# 选择一张测试图
if TEST_IMAGE_PATH is not None:
    sample_image = Path(TEST_IMAGE_PATH)
elif PLAYER_DATASET_DIR is not None and (PLAYER_DATASET_DIR / "valid" / "images").exists():
    candidates = list(iter_images(PLAYER_DATASET_DIR / "valid" / "images"))
    sample_image = candidates[0] if candidates else None
else:
    sample_image = None

if sample_image is None:
    print("请设置 TEST_IMAGE_PATH，或确保 player 数据集 valid/images 存在。")
else:
    _, pipeline_records = run_pipeline(sample_image, show=True)
````
